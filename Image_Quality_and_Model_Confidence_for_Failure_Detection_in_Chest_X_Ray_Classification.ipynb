```python
# ============================================================
# COVID-19 RADIOGRAPHY DATABASE
# COMPLETE DATASET AUDIT — FIXED VERSION
# ============================================================

import os
import zipfile
import hashlib
import random
import shutil
import warnings
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
from PIL import Image, ImageFile
import matplotlib.pyplot as plt

from IPython.display import display

ImageFile.LOAD_TRUNCATED_IMAGES = False
warnings.filterwarnings("ignore")


# ============================================================
# 1. CONFIGURATION
# ============================================================

ZIP_PATH = Path(
    r"C:\Users\mohsi\Downloads\archive (1).zip"
)

EXTRACT_DIR = Path(
    r"C:\Users\mohsi\Downloads\covid_radiography_dataset"
)

REPORT_DIR = EXTRACT_DIR / "audit_reports"

IMAGE_EXTENSIONS = {
    ".png",
    ".jpg",
    ".jpeg",
    ".bmp",
    ".tif",
    ".tiff"
}


# ============================================================
# 2. START
# ============================================================

print("=" * 80)
print("COVID-19 RADIOGRAPHY DATABASE — COMPLETE DATASET AUDIT")
print("=" * 80)

if not ZIP_PATH.exists():

    raise FileNotFoundError(
        f"\nZIP FILE NOT FOUND:\n{ZIP_PATH}"
    )

zip_size_gb = ZIP_PATH.stat().st_size / (1024 ** 3)

print(f"\nZIP file:")
print(ZIP_PATH)

print(f"\nZIP size:")
print(f"{zip_size_gb:.2f} GB")


# ============================================================
# 3. INSPECT ZIP
# ============================================================

print("\n" + "=" * 80)
print("1. ZIP CONTENT INSPECTION")
print("=" * 80)

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    zip_entries = z.infolist()

print(
    f"Total ZIP entries: {len(zip_entries):,}"
)

zip_rows = []

for item in zip_entries:

    zip_rows.append({
        "path": item.filename,
        "size_MB": item.file_size / (1024 ** 2),
        "compressed_MB": item.compress_size / (1024 ** 2),
        "is_directory": item.is_dir()
    })

zip_df = pd.DataFrame(zip_rows)

print("\nFirst 30 ZIP entries:")

display(
    zip_df.head(30)
)

print("\nLargest ZIP entries:")

display(
    zip_df
    .sort_values(
        "size_MB",
        ascending=False
    )
    .head(20)
)


# ============================================================
# 4. REMOVE OLD / BROKEN EXTRACTION
# ============================================================

print("\n" + "=" * 80)
print("2. PREPARING EXTRACTION DIRECTORY")
print("=" * 80)

if EXTRACT_DIR.exists():

    print(
        "Existing extraction directory found."
    )

    print(
        "Removing it to guarantee a clean extraction..."
    )

    shutil.rmtree(EXTRACT_DIR)

EXTRACT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

REPORT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# 5. EXTRACT ZIP
# ============================================================

print("\n" + "=" * 80)
print("3. EXTRACTING DATASET")
print("=" * 80)

print(
    "Extraction started..."
)

with zipfile.ZipFile(ZIP_PATH, "r") as z:

    z.extractall(EXTRACT_DIR)

print(
    "Extraction completed successfully."
)


# ============================================================
# 6. FIND ACTUAL DATASET ROOT
# ============================================================

print("\n" + "=" * 80)
print("4. FINDING DATASET ROOT")
print("=" * 80)

dataset_root = None

possible_root = (
    EXTRACT_DIR /
    "COVID-19_Radiography_Dataset"
)

if possible_root.exists():

    dataset_root = possible_root

else:

    # Search for the folder automatically
    for p in EXTRACT_DIR.rglob("*"):

        if p.is_dir() and (
            p.name.lower()
            == "covid-19_radiography_dataset".lower()
        ):

            dataset_root = p
            break


if dataset_root is None:

    print(
        "❌ Could not automatically find dataset root."
    )

    print("\nExtracted folders:")

    for p in EXTRACT_DIR.rglob("*"):

        if p.is_dir():

            print(
                p.relative_to(EXTRACT_DIR)
            )

    raise RuntimeError(
        "Dataset root could not be found."
    )


print(
    f"Dataset root:\n{dataset_root}"
)


# ============================================================
# 7. DIRECTORY STRUCTURE
# ============================================================

print("\n" + "=" * 80)
print("5. DATASET DIRECTORY STRUCTURE")
print("=" * 80)

for p in sorted(dataset_root.iterdir()):

    if p.is_dir():

        print(
            f"📁 {p.name}"
        )

    else:

        print(
            f"📄 {p.name}"
        )


# ============================================================
# 8. FIND ALL IMAGES
# ============================================================

print("\n" + "=" * 80)
print("6. IMAGE DISCOVERY")
print("=" * 80)

image_files = []

for p in dataset_root.rglob("*"):

    if not p.is_file():
        continue

    if p.suffix.lower() not in IMAGE_EXTENSIONS:
        continue

    if REPORT_DIR in p.parents:
        continue

    image_files.append(p)


print(
    f"Total image files found: {len(image_files):,}"
)


# ============================================================
# 9. IMAGE COUNT BY FOLDER
# ============================================================

print("\n" + "=" * 80)
print("7. IMAGES BY CLASS/FOLDER")
print("=" * 80)

folder_counts = Counter()

for p in image_files:

    try:

        relative = p.relative_to(dataset_root)

        if len(relative.parts) >= 2:

            class_name = relative.parts[0]

        else:

            class_name = "UNKNOWN"

        folder_counts[class_name] += 1

    except:

        folder_counts["UNKNOWN"] += 1


for folder, count in folder_counts.most_common():

    print(
        f"{folder:30} {count:,}"
    )


# ============================================================
# 10. IMAGE VALIDITY + BASIC METADATA
# ============================================================

print("\n" + "=" * 80)
print("8. IMAGE VALIDITY + METADATA")
print("=" * 80)

records = []
corrupted = []

total_images = len(image_files)

for i, path in enumerate(image_files):

    if i % 1000 == 0:

        print(
            f"Processing {i:,}/{total_images:,}"
        )

    try:

        with Image.open(path) as img:

            img.load()

            width, height = img.size

            mode = img.mode

            image_format = img.format

            gray = np.asarray(
                img.convert("L"),
                dtype=np.float32
            )

            mean_brightness = float(
                gray.mean()
            )

            std_brightness = float(
                gray.std()
            )

            min_pixel = int(
                gray.min()
            )

            max_pixel = int(
                gray.max()
            )


        relative = path.relative_to(
            dataset_root
        )

        if len(relative.parts) >= 2:

            class_name = relative.parts[0]

        else:

            class_name = "UNKNOWN"


        records.append({

            "path": str(path),

            "relative_path":
                str(relative),

            "filename":
                path.name,

            "class":
                class_name,

            "extension":
                path.suffix.lower(),

            "format":
                image_format,

            "width":
                width,

            "height":
                height,

            "mode":
                mode,

            "file_size_MB":
                path.stat().st_size /
                (1024 ** 2),

            "mean_brightness":
                mean_brightness,

            "std_brightness":
                std_brightness,

            "min_pixel":
                min_pixel,

            "max_pixel":
                max_pixel

        })

    except Exception as e:

        corrupted.append({

            "path":
                str(path),

            "filename":
                path.name,

            "error":
                str(e)

        })


metadata_df = pd.DataFrame(
    records
)

corrupt_df = pd.DataFrame(
    corrupted
)


print("\n" + "-" * 60)

print(
    f"Readable images:  {len(metadata_df):,}"
)

print(
    f"Corrupted images: {len(corrupted):,}"
)


if len(corrupt_df) > 0:

    print("\nFirst corrupted images:")

    display(
        corrupt_df.head(20)
    )


# ============================================================
# 11. IMAGE DIMENSIONS
# ============================================================

print("\n" + "=" * 80)
print("9. IMAGE DIMENSIONS")
print("=" * 80)

if len(metadata_df) > 0:

    dimension_counts = (
        metadata_df
        .groupby(
            ["width", "height"]
        )
        .size()
        .sort_values(
            ascending=False
        )
    )

    print(
        f"Unique dimensions: "
        f"{len(dimension_counts):,}"
    )

    display(
        dimension_counts
        .head(30)
        .reset_index(
            name="count"
        )
    )


# ============================================================
# 12. IMAGE MODES
# ============================================================

print("\n" + "=" * 80)
print("10. IMAGE MODES")
print("=" * 80)

if len(metadata_df) > 0:

    display(
        metadata_df["mode"]
        .value_counts()
        .rename_axis("mode")
        .reset_index(
            name="count"
        )
    )


# ============================================================
# 13. IMAGE FORMATS
# ============================================================

print("\n" + "=" * 80)
print("11. IMAGE FORMATS")
print("=" * 80)

if len(metadata_df) > 0:

    display(
        metadata_df["format"]
        .value_counts()
        .rename_axis("format")
        .reset_index(
            name="count"
        )
    )


# ============================================================
# 14. CLASS DISTRIBUTION
# ============================================================

print("\n" + "=" * 80)
print("12. CLASS DISTRIBUTION")
print("=" * 80)

if len(metadata_df) > 0:

    class_counts = (
        metadata_df["class"]
        .value_counts()
    )

    display(
        class_counts
        .rename_axis("class")
        .reset_index(
            name="images"
        )
    )

    plt.figure(
        figsize=(10, 6)
    )

    class_counts.plot(
        kind="bar"
    )

    plt.title(
        "COVID-19 Radiography Dataset — Class Distribution"
    )

    plt.xlabel(
        "Class"
    )

    plt.ylabel(
        "Number of Images"
    )

    plt.xticks(
        rotation=45
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 15. EXACT DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("13. EXACT DUPLICATE ANALYSIS")
print("=" * 80)

def calculate_md5(path):

    md5 = hashlib.md5()

    with open(
        path,
        "rb"
    ) as f:

        while True:

            chunk = f.read(
                1024 * 1024
            )

            if not chunk:
                break

            md5.update(
                chunk
            )

    return md5.hexdigest()


hash_records = []

total_readable = len(metadata_df)

for i, path in enumerate(
    metadata_df["path"]
):

    if i % 1000 == 0:

        print(
            f"Hashing "
            f"{i:,}/{total_readable:,}"
        )

    try:

        hash_records.append({

            "path":
                path,

            "hash":
                calculate_md5(path)

        })

    except Exception as e:

        print(
            "Hash error:",
            path,
            e
        )


hash_df = pd.DataFrame(
    hash_records
)


if len(hash_df) > 0:

    unique_hashes = (
        hash_df["hash"]
        .nunique()
    )

    duplicate_file_count = (
        len(hash_df)
        - unique_hashes
    )

    duplicate_groups = (
        hash_df
        .groupby("hash")
        .size()
    )

    duplicate_groups = (
        duplicate_groups[
            duplicate_groups > 1
        ]
    )

else:

    unique_hashes = 0

    duplicate_file_count = 0

    duplicate_groups = pd.Series(
        dtype=int
    )


print("\nHASH RESULTS")

print(
    f"Readable images: "
    f"{len(hash_df):,}"
)

print(
    f"Unique hashes: "
    f"{unique_hashes:,}"
)

print(
    f"Duplicate files: "
    f"{duplicate_file_count:,}"
)

print(
    f"Duplicate groups: "
    f"{len(duplicate_groups):,}"
)


# ============================================================
# 16. CROSS-FOLDER DUPLICATES
# ============================================================

print("\n" + "=" * 80)
print("14. CROSS-FOLDER DUPLICATE ANALYSIS")
print("=" * 80)

cross_folder_duplicates = []

if len(hash_df) > 0:

    hash_df["class"] = hash_df[
        "path"
    ].apply(
        lambda x:
        Path(x)
        .relative_to(dataset_root)
        .parts[0]
    )

    for h, group in (
        hash_df.groupby("hash")
    ):

        classes = (
            group["class"]
            .unique()
        )

        if len(classes) > 1:

            cross_folder_duplicates.append({

                "hash":
                    h,

                "number_of_files":
                    len(group),

                "classes":
                    list(classes),

                "paths":
                    list(group["path"])

            })


print(
    "Duplicate hashes appearing "
    "across different classes/folders:"
)

print(
    len(cross_folder_duplicates)
)


if cross_folder_duplicates:

    display(
        pd.DataFrame(
            cross_folder_duplicates
        )
        .head(30)
    )


# ============================================================
# 17. METADATA FILES
# ============================================================

print("\n" + "=" * 80)
print("15. METADATA FILES")
print("=" * 80)

metadata_files = []

for p in dataset_root.rglob("*"):

    if not p.is_file():
        continue

    if p.suffix.lower() in {
        ".csv",
        ".xlsx",
        ".json",
        ".txt"
    }:

        metadata_files.append(p)


print(
    f"Metadata files found: "
    f"{len(metadata_files):,}"
)


for p in metadata_files:

    print(
        "\n📄",
        p.relative_to(dataset_root)
    )

    try:

        if p.suffix.lower() == ".csv":

            temp = pd.read_csv(
                p,
                nrows=5
            )

            print(
                "Columns:",
                list(temp.columns)
            )

            display(temp)

        elif p.suffix.lower() == ".xlsx":

            excel = pd.ExcelFile(
                p
            )

            print(
                "Sheets:",
                excel.sheet_names
            )

            for sheet in excel.sheet_names:

                temp = pd.read_excel(
                    p,
                    sheet_name=sheet,
                    nrows=5
                )

                print(
                    f"\nSheet: {sheet}"
                )

                print(
                    "Columns:",
                    list(temp.columns)
                )

                display(temp)

        elif p.suffix.lower() == ".txt":

            with open(
                p,
                "r",
                encoding="utf-8",
                errors="ignore"
            ) as f:

                print(
                    f.read(2000)
                )

    except Exception as e:

        print(
            "Could not read:",
            e
        )


# ============================================================
# 18. PATIENT / SOURCE IDENTIFIER SEARCH
# ============================================================

print("\n" + "=" * 80)
print("16. PATIENT / SOURCE IDENTIFIER SEARCH")
print("=" * 80)

identifier_keywords = [

    "patient",
    "patient_id",
    "patientid",

    "subject",
    "subject_id",

    "study",
    "study_id",

    "image",
    "image_id",

    "uid",

    "source",

    "dataset"

]

identifier_hits = []


for p in metadata_files:

    try:

        if p.suffix.lower() == ".csv":

            temp = pd.read_csv(
                p,
                nrows=5
            )

            columns = temp.columns

        elif p.suffix.lower() == ".xlsx":

            excel = pd.ExcelFile(
                p
            )

            columns = []

            for sheet in excel.sheet_names:

                temp = pd.read_excel(
                    p,
                    sheet_name=sheet,
                    nrows=5
                )

                columns.extend(
                    list(temp.columns)
                )

        else:

            continue


        for col in columns:

            col_lower = str(
                col
            ).lower()

            for keyword in identifier_keywords:

                if keyword in col_lower:

                    identifier_hits.append({

                        "file":
                            str(
                                p.relative_to(
                                    dataset_root
                                )
                            ),

                        "column":
                            str(col),

                        "keyword":
                            keyword

                    })


    except:

        pass


if identifier_hits:

    display(
        pd.DataFrame(
            identifier_hits
        )
        .drop_duplicates()
    )

else:

    print(
        "⚠️ No obvious patient/study "
        "identifiers found."
    )


# ============================================================
# 19. IMAGE QUALITY STATISTICS
# ============================================================

print("\n" + "=" * 80)
print("17. IMAGE QUALITY STATISTICS")
print("=" * 80)

if len(metadata_df) > 0:

    display(
        metadata_df[
            [
                "file_size_MB",
                "width",
                "height",
                "mean_brightness",
                "std_brightness",
                "min_pixel",
                "max_pixel"
            ]
        ].describe()
    )


# ============================================================
# 20. DARK / BRIGHT IMAGES
# ============================================================

print("\n" + "=" * 80)
print("18. EXTREME IMAGE EXAMPLES")
print("=" * 80)

if len(metadata_df) > 0:

    print("\nDarkest images:")

    display(
        metadata_df
        .sort_values(
            "mean_brightness"
        )
        [
            [
                "relative_path",
                "class",
                "mean_brightness",
                "std_brightness",
                "width",
                "height"
            ]
        ]
        .head(10)
    )

    print("\nBrightest images:")

    display(
        metadata_df
        .sort_values(
            "mean_brightness",
            ascending=False
        )
        [
            [
                "relative_path",
                "class",
                "mean_brightness",
                "std_brightness",
                "width",
                "height"
            ]
        ]
        .head(10)
    )


# ============================================================
# 21. RANDOM IMAGE VISUALIZATION
# ============================================================

print("\n" + "=" * 80)
print("19. RANDOM IMAGE INSPECTION")
print("=" * 80)

if len(metadata_df) > 0:

    random.seed(42)

    sample_paths = random.sample(

        metadata_df["path"].tolist(),

        min(
            16,
            len(metadata_df)
        )

    )

    fig, axes = plt.subplots(
        4,
        4,
        figsize=(16, 16)
    )

    for ax in axes.flat:

        ax.axis("off")


    for ax, path in zip(
        axes.flat,
        sample_paths
    ):

        try:

            img = Image.open(
                path
            )

            ax.imshow(
                img,
                cmap="gray"
            )

            ax.set_title(
                Path(path).parent.name,
                fontsize=9
            )

            ax.axis("off")

        except:

            pass


    plt.tight_layout()

    plt.show()


# ============================================================
# 22. ONE VISUAL GRID PER CLASS
# ============================================================

print("\n" + "=" * 80)
print("20. CLASS VISUAL INSPECTION")
print("=" * 80)

if len(metadata_df) > 0:

    for class_name in (
        metadata_df["class"]
        .unique()
    ):

        class_df = metadata_df[
            metadata_df["class"]
            == class_name
        ]

        if len(class_df) == 0:
            continue


        samples = class_df.sample(

            min(
                8,
                len(class_df)
            ),

            random_state=42

        )


        fig, axes = plt.subplots(
            2,
            4,
            figsize=(16, 8)
        )


        fig.suptitle(
            f"CLASS: {class_name}",
            fontsize=16
        )


        for ax in axes.flat:

            ax.axis("off")


        for ax, (_, row) in zip(
            axes.flat,
            samples.iterrows()
        ):

            try:

                img = Image.open(
                    row["path"]
                )

                ax.imshow(
                    img,
                    cmap="gray"
                )

                ax.set_title(
                    Path(
                        row["path"]
                    ).name,
                    fontsize=7
                )

                ax.axis("off")

            except:

                pass


        plt.tight_layout()

        plt.show()


# ============================================================
# 23. SAVE REPORTS
# ============================================================

print("\n" + "=" * 80)
print("21. SAVING AUDIT REPORTS")
print("=" * 80)

metadata_path = (
    REPORT_DIR /
    "image_metadata.csv"
)

hash_path = (
    REPORT_DIR /
    "image_hashes.csv"
)

corrupt_path = (
    REPORT_DIR /
    "corrupted_images.csv"
)

zip_path = (
    REPORT_DIR /
    "zip_contents.csv"
)


metadata_df.to_csv(
    metadata_path,
    index=False
)

hash_df.to_csv(
    hash_path,
    index=False
)

corrupt_df.to_csv(
    corrupt_path,
    index=False
)

zip_df.to_csv(
    zip_path,
    index=False
)


print(
    "Saved:"
)

print(
    metadata_path
)

print(
    hash_path
)

print(
    corrupt_path
)

print(
    zip_path
)


# ============================================================
# 24. FINAL DATASET REPORT
# ============================================================

print("\n")
print("=" * 80)
print("FINAL DATASET REPORT")
print("=" * 80)

print(
    f"""
Dataset root:
{dataset_root}

ZIP:
{ZIP_PATH}

ZIP size:
{zip_size_gb:.2f} GB

Total image files:
{len(image_files):,}

Readable images:
{len(metadata_df):,}

Corrupted images:
{len(corrupted):,}

Unique dimensions:
{
    metadata_df[
        ["width", "height"]
    ].drop_duplicates().shape[0]
    if len(metadata_df) > 0
    else 0
:,}

Unique image hashes:
{unique_hashes:,}

Duplicate files:
{duplicate_file_count:,}

Duplicate groups:
{len(duplicate_groups):,}

Cross-class duplicate hashes:
{len(cross_folder_duplicates):,}

Classes:
{
    class_counts.to_string()
    if len(metadata_df) > 0
    else "No classes detected."
}
"""
)

print("=" * 80)
print("AUDIT COMPLETE")
print("=" * 80)


# ============================================================
# 25. FINAL WARNING
# ============================================================

print(
    """
DO NOT TRAIN ANY MODEL YET.

We need to inspect the results first.

Send me these sections:

1. FINAL DATASET REPORT
2. IMAGES BY CLASS/FOLDER
3. EXACT DUPLICATE ANALYSIS
4. CROSS-FOLDER DUPLICATES
5. PATIENT / SOURCE IDENTIFIER SEARCH
6. IMAGE DIMENSIONS
7. RANDOM IMAGE GRID if anything looks suspicious

Then we decide:

    ✅ USE THIS DATASET

or

    ❌ REJECT THIS DATASET
"""
)
```

```python
from pathlib import Path
import pandas as pd
import hashlib
from PIL import Image
from sklearn.model_selection import train_test_split
```

```python
DATASET_ROOT = Path(
    r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset"
)

OUTPUT_DIR = DATASET_ROOT / "research_dataset"
OUTPUT_DIR.mkdir(exist_ok=True)

print("Dataset root:")
print(DATASET_ROOT)

print("\nOutput directory:")
print(OUTPUT_DIR)
```

```python
# ------------------------------------------------------------
# EXACT DUPLICATE DETECTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CHECKING EXACT DUPLICATES")
print("=" * 70)

def file_md5(path, chunk_size=1024 * 1024):

    md5 = hashlib.md5()

    with open(path, "rb") as f:

        while True:

            chunk = f.read(chunk_size)

            if not chunk:
                break

            md5.update(chunk)

    return md5.hexdigest()


hashes = []

for i, path in enumerate(df["path"]):

    hashes.append(file_md5(path))

    if (i + 1) % 1000 == 0:
        print(f"Hashed {i + 1:,}/{len(df):,}")

df["md5"] = hashes

duplicate_mask = df.duplicated(
    subset=["md5"],
    keep="first"
)

duplicates = df[duplicate_mask].copy()

print("\nUnique files:", df["md5"].nunique())
print("Duplicate files:", len(duplicates))
```

```python
# ============================================================
# STEP 5 — PYTORCH DATASET + DATALOADERS
# ============================================================

import pandas as pd
import torch
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
import os

print("=" * 70)
print("BUILDING PYTORCH DATASET")
print("=" * 70)

# ------------------------------------------------------------
# 1. DEVICE
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print(f"\nDevice: {device}")

# ------------------------------------------------------------
# 2. LOAD CSV FILES
# ------------------------------------------------------------

train_df = pd.read_csv(
    OUTPUT_DIR / "train.csv"
)

val_df = pd.read_csv(
    OUTPUT_DIR / "validation.csv"
)

test_df = pd.read_csv(
    OUTPUT_DIR / "test.csv"
)

print("\nLoaded:")
print(f"Train      : {len(train_df):,}")
print(f"Validation : {len(val_df):,}")
print(f"Test       : {len(test_df):,}")

# ------------------------------------------------------------
# 3. TRANSFORMS
# ------------------------------------------------------------
#
# DenseNet121 expects 3-channel ImageNet-style input.
# Our X-rays are grayscale, so we convert them to RGB.
#
# Training gets mild augmentation.
# Validation/test remain deterministic.
# ------------------------------------------------------------

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=5
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# ------------------------------------------------------------
# 4. CUSTOM DATASET
# ------------------------------------------------------------

class ChestXrayDataset(Dataset):

    def __init__(
        self,
        dataframe,
        transform=None
    ):

        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):

        return len(self.df)

    def __getitem__(self, index):

        row = self.df.iloc[index]

        image_path = row["path"]
        label = int(row["label"])

        # Load grayscale image
        image = Image.open(
            image_path
        ).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        return image, label


# ------------------------------------------------------------
# 5. CREATE DATASETS
# ------------------------------------------------------------

train_dataset = ChestXrayDataset(
    train_df,
    transform=train_transform
)

val_dataset = ChestXrayDataset(
    val_df,
    transform=eval_transform
)

test_dataset = ChestXrayDataset(
    test_df,
    transform=eval_transform
)

print("\nDatasets created:")
print(f"Train dataset      : {len(train_dataset):,}")
print(f"Validation dataset : {len(val_dataset):,}")
print(f"Test dataset       : {len(test_dataset):,}")

# ------------------------------------------------------------
# 6. DATALOADERS
# ------------------------------------------------------------

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=True
)

print("\nDataLoaders created:")
print(f"Batch size         : {BATCH_SIZE}")
print(f"Train batches      : {len(train_loader):,}")
print(f"Validation batches : {len(val_loader):,}")
print(f"Test batches       : {len(test_loader):,}")

# ------------------------------------------------------------
# 7. TEST ONE BATCH
# ------------------------------------------------------------

images, labels = next(iter(train_loader))

print("\n" + "=" * 70)
print("BATCH TEST")
print("=" * 70)

print(f"Image tensor shape : {images.shape}")
print(f"Label shape        : {labels.shape}")
print(f"Image dtype        : {images.dtype}")
print(f"Label dtype        : {labels.dtype}")

print(
    f"Image device       : {images.device}"
)

print(
    f"Pixel range        : "
    f"{images.min().item():.3f} → "
    f"{images.max().item():.3f}"
)

print(
    f"Labels in batch    : "
    f"{labels.tolist()}"
)

# ------------------------------------------------------------
# 8. MOVE BATCH TO GPU
# ------------------------------------------------------------

if torch.cuda.is_available():

    images_gpu = images.to(
        device,
        non_blocking=True
    )

    labels_gpu = labels.to(
        device,
        non_blocking=True
    )

    print("\nGPU batch test: SUCCESS")

    print(
        f"GPU image shape   : {images_gpu.shape}"
    )

    print(
        f"GPU labels shape  : {labels_gpu.shape}"
    )

    del images_gpu, labels_gpu

    torch.cuda.empty_cache()

print("\n" + "=" * 70)
print("✅ DATALOADER SETUP COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# STEP 6 — DENSENET121 BASELINE MODEL
# ============================================================

import torch
import torch.nn as nn
from torchvision.models import densenet121, DenseNet121_Weights

print("=" * 70)
print("BUILDING DENSENET121 BASELINE")
print("=" * 70)

# ------------------------------------------------------------
# 1. DEVICE
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print(f"\nDevice: {device}")

# ------------------------------------------------------------
# 2. LOAD PRETRAINED DENSENET121
# ------------------------------------------------------------

weights = DenseNet121_Weights.DEFAULT

model = densenet121(
    weights=weights
)

# ------------------------------------------------------------
# 3. REPLACE CLASSIFIER
# ------------------------------------------------------------

num_features = model.classifier.in_features

model.classifier = nn.Linear(
    num_features,
    2
)

# ------------------------------------------------------------
# 4. MOVE TO GPU
# ------------------------------------------------------------

model = model.to(device)

# ------------------------------------------------------------
# 5. CLASS WEIGHTS
# ------------------------------------------------------------
#
# Normal       = 10,191
# LungOpacity  =  6,012
#
# Weight is inversely proportional to class frequency.
# ------------------------------------------------------------

class_counts = torch.tensor([
    (train_df["label"] == 0).sum(),
    (train_df["label"] == 1).sum()
], dtype=torch.float32)

class_weights = (
    class_counts.sum()
    /
    (2 * class_counts)
)

class_weights = class_weights.to(device)

print("\nClass counts:")
print(f"Normal       : {int(class_counts[0])}")
print(f"Lung_Opacity : {int(class_counts[1])}")

print("\nClass weights:")
print(
    f"Normal       : {class_weights[0].item():.4f}"
)

print(
    f"Lung_Opacity : {class_weights[1].item():.4f}"
)

# ------------------------------------------------------------
# 6. LOSS FUNCTION
# ------------------------------------------------------------

criterion = nn.CrossEntropyLoss(
    weight=class_weights
)

# ------------------------------------------------------------
# 7. OPTIMIZER
# ------------------------------------------------------------

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=1e-4,
    weight_decay=1e-4
)

# ------------------------------------------------------------
# 8. MODEL INFORMATION
# ------------------------------------------------------------

total_params = sum(
    p.numel()
    for p in model.parameters()
)

trainable_params = sum(
    p.numel()
    for p in model.parameters()
    if p.requires_grad
)

print("\n" + "=" * 70)
print("MODEL INFORMATION")
print("=" * 70)

print(f"Architecture       : DenseNet121")
print(f"Pretrained         : ImageNet")
print(f"Input size         : 224 × 224")
print(f"Output classes     : 2")
print(f"Total parameters   : {total_params:,}")
print(f"Trainable params   : {trainable_params:,}")
print(f"Learning rate      : 1e-4")
print(f"Weight decay       : 1e-4")
print(f"Loss               : Weighted Cross Entropy")

# ------------------------------------------------------------
# 9. FORWARD PASS TEST
# ------------------------------------------------------------

model.eval()

with torch.no_grad():

    sample_images = images.to(device)

    outputs = model(
        sample_images
    )

print("\n" + "=" * 70)
print("FORWARD PASS TEST")
print("=" * 70)

print(
    f"Input shape        : {sample_images.shape}"
)

print(
    f"Output shape       : {outputs.shape}"
)

print(
    f"Output example     : {outputs[0].detach().cpu().numpy()}"
)

# ------------------------------------------------------------
# 10. CLEAN GPU MEMORY
# ------------------------------------------------------------

del sample_images, outputs

torch.cuda.empty_cache()

print("\n" + "=" * 70)
print("✅ DENSENET121 READY")
print("=" * 70)
```

```python
# ============================================================
# STEP 7 — TRAIN DENSENET121 CLEAN BASELINE
# ============================================================

import time
import copy
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("=" * 70)
print("TRAINING DENSENET121")
print("=" * 70)

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

NUM_EPOCHS = 10
PATIENCE = 3

BEST_MODEL_PATH = OUTPUT_DIR / "densenet121_best.pth"

print(f"\nMaximum epochs : {NUM_EPOCHS}")
print(f"Early stopping : {PATIENCE} epochs")
print(f"Checkpoint     : {BEST_MODEL_PATH}")

# ------------------------------------------------------------
# EVALUATION FUNCTION
# ------------------------------------------------------------

def evaluate_model(model, loader, criterion, device):

    model.eval()

    running_loss = 0.0

    all_labels = []
    all_probs = []
    all_preds = []

    with torch.no_grad():

        for batch_images, batch_labels in loader:

            batch_images = batch_images.to(
                device,
                non_blocking=True
            )

            batch_labels = batch_labels.to(
                device,
                non_blocking=True
            )

            outputs = model(batch_images)

            loss = criterion(
                outputs,
                batch_labels
            )

            probabilities = torch.softmax(
                outputs,
                dim=1
            )[:, 1]

            predictions = (
                probabilities >= 0.5
            ).long()

            running_loss += (
                loss.item()
                * batch_images.size(0)
            )

            all_labels.extend(
                batch_labels.cpu().numpy()
            )

            all_probs.extend(
                probabilities.cpu().numpy()
            )

            all_preds.extend(
                predictions.cpu().numpy()
            )

    avg_loss = (
        running_loss / len(loader.dataset)
    )

    all_labels = np.array(all_labels)
    all_probs = np.array(all_probs)
    all_preds = np.array(all_preds)

    accuracy = accuracy_score(
        all_labels,
        all_preds
    )

    precision = precision_score(
        all_labels,
        all_preds,
        zero_division=0
    )

    recall = recall_score(
        all_labels,
        all_preds,
        zero_division=0
    )

    f1 = f1_score(
        all_labels,
        all_preds,
        zero_division=0
    )

    auroc = roc_auc_score(
        all_labels,
        all_probs
    )

    return {
        "loss": avg_loss,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "auroc": auroc
    }


# ------------------------------------------------------------
# TRAINING
# ------------------------------------------------------------

history = []

best_auroc = -np.inf
best_epoch = 0
epochs_without_improvement = 0

best_state = None

training_start = time.time()

for epoch in range(1, NUM_EPOCHS + 1):

    epoch_start = time.time()

    model.train()

    running_loss = 0.0

    all_train_labels = []
    all_train_probs = []

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    for batch_images, batch_labels in train_loader:

        batch_images = batch_images.to(
            device,
            non_blocking=True
        )

        batch_labels = batch_labels.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        outputs = model(
            batch_images
        )

        loss = criterion(
            outputs,
            batch_labels
        )

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item()
            * batch_images.size(0)
        )

        probabilities = torch.softmax(
            outputs.detach(),
            dim=1
        )[:, 1]

        all_train_labels.extend(
            batch_labels.detach().cpu().numpy()
        )

        all_train_probs.extend(
            probabilities.cpu().numpy()
        )

    train_loss = (
        running_loss /
        len(train_loader.dataset)
    )

    # --------------------------------------------------------
    # TRAIN AUROC
    # --------------------------------------------------------

    train_auroc = roc_auc_score(
        all_train_labels,
        all_train_probs
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    val_metrics = evaluate_model(
        model,
        val_loader,
        criterion,
        device
    )

    epoch_time = (
        time.time() - epoch_start
    )

    # --------------------------------------------------------
    # SAVE HISTORY
    # --------------------------------------------------------

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "train_auroc": train_auroc,
        "val_loss": val_metrics["loss"],
        "val_accuracy": val_metrics["accuracy"],
        "val_precision": val_metrics["precision"],
        "val_recall": val_metrics["recall"],
        "val_f1": val_metrics["f1"],
        "val_auroc": val_metrics["auroc"],
        "epoch_time_sec": epoch_time
    })

    # --------------------------------------------------------
    # PRINT
    # --------------------------------------------------------

    print(
        f"\nEpoch {epoch:02d}/{NUM_EPOCHS}"
    )

    print(
        f"  Train Loss   : {train_loss:.4f}"
    )

    print(
        f"  Train AUROC  : {train_auroc:.4f}"
    )

    print(
        f"  Val Loss     : {val_metrics['loss']:.4f}"
    )

    print(
        f"  Val Accuracy : {val_metrics['accuracy']:.4f}"
    )

    print(
        f"  Val F1       : {val_metrics['f1']:.4f}"
    )

    print(
        f"  Val AUROC    : {val_metrics['auroc']:.4f}"
    )

    print(
        f"  Time         : {epoch_time:.1f}s"
    )

    # --------------------------------------------------------
    # BEST MODEL
    # --------------------------------------------------------

    if val_metrics["auroc"] > best_auroc:

        best_auroc = val_metrics["auroc"]

        best_epoch = epoch

        epochs_without_improvement = 0

        best_state = copy.deepcopy(
            model.state_dict()
        )

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "val_auroc": best_auroc
            },
            BEST_MODEL_PATH
        )

        print(
            f"  ⭐ NEW BEST MODEL "
            f"(Val AUROC = {best_auroc:.4f})"
        )

    else:

        epochs_without_improvement += 1

        print(
            f"  No improvement "
            f"({epochs_without_improvement}/{PATIENCE})"
        )

    # --------------------------------------------------------
    # EARLY STOPPING
    # --------------------------------------------------------

    if epochs_without_improvement >= PATIENCE:

        print(
            "\n⏹ Early stopping triggered."
        )

        break

# ------------------------------------------------------------
# RESTORE BEST MODEL
# ------------------------------------------------------------

if best_state is not None:

    model.load_state_dict(
        best_state
    )

training_time = (
    time.time() - training_start
)

# ------------------------------------------------------------
# SAVE HISTORY
# ------------------------------------------------------------

history_df = pd.DataFrame(history)

history_path = (
    OUTPUT_DIR /
    "training_history.csv"
)

history_df.to_csv(
    history_path,
    index=False
)

# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("✅ TRAINING COMPLETE")
print("=" * 70)

print(
    f"\nBest epoch       : {best_epoch}"
)

print(
    f"Best Val AUROC   : {best_auroc:.4f}"
)

print(
    f"Training time    : "
    f"{training_time / 60:.2f} minutes"
)

print(
    f"Model saved      : {BEST_MODEL_PATH}"
)

print(
    f"History saved    : {history_path}"
)

# ------------------------------------------------------------
# GPU MEMORY
# ------------------------------------------------------------

if torch.cuda.is_available():

    allocated = (
        torch.cuda.memory_allocated()
        / (1024 ** 3)
    )

    reserved = (
        torch.cuda.memory_reserved()
        / (1024 ** 3)
    )

    print(
        f"\nGPU memory allocated : "
        f"{allocated:.2f} GB"
    )

    print(
        f"GPU memory reserved  : "
        f"{reserved:.2f} GB"
    )
```

```python
# ================================================================
# STEP 9 — CREATE CONTROLLED DEGRADATION TEST SETS
# ================================================================

import os
import cv2
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm

print("=" * 70)
print("CREATING CONTROLLED DEGRADATION TEST SETS")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

RESEARCH_DIR = r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset\research_dataset"

TEST_CSV = os.path.join(
    RESEARCH_DIR,
    "test.csv"
)

DEGRADATION_ROOT = os.path.join(
    RESEARCH_DIR,
    "degradation_test_sets"
)

os.makedirs(DEGRADATION_ROOT, exist_ok=True)

print(f"Test CSV: {TEST_CSV}")
print(f"Output:   {DEGRADATION_ROOT}")


# ------------------------------------------------
# DEGRADATION FUNCTIONS
# ------------------------------------------------

def gaussian_blur(img, strength=3):
    return cv2.GaussianBlur(
        img,
        (strength, strength),
        0
    )


def gaussian_noise(img, std=20):
    noise = np.random.normal(
        0,
        std,
        img.shape
    )

    noisy = img.astype(np.float32) + noise

    return np.clip(
        noisy,
        0,
        255
    ).astype(np.uint8)


def low_resolution(img, size=64):
    small = cv2.resize(
        img,
        (size, size),
        interpolation=cv2.INTER_AREA
    )

    return cv2.resize(
        small,
        (299, 299),
        interpolation=cv2.INTER_LINEAR
    )


def jpeg_compression(img, quality=30):
    encode_param = [
        int(cv2.IMWRITE_JPEG_QUALITY),
        quality
    ]

    success, encoded = cv2.imencode(
        ".jpg",
        img,
        encode_param
    )

    if not success:
        return img

    return cv2.imdecode(
        encoded,
        cv2.IMREAD_GRAYSCALE
    )


def brightness_contrast(img, alpha=0.65, beta=-25):
    result = img.astype(np.float32) * alpha + beta

    return np.clip(
        result,
        0,
        255
    ).astype(np.uint8)


# ------------------------------------------------
# DEGRADATIONS
# ------------------------------------------------

DEGRADATIONS = {
    "blur": lambda img: gaussian_blur(img, 3),
    "noise": lambda img: gaussian_noise(img, 20),
    "low_resolution": lambda img: low_resolution(img, 64),
    "jpeg": lambda img: jpeg_compression(img, 30),
    "brightness_contrast": lambda img: brightness_contrast(
        img,
        0.65,
        -25
    )
}


# ------------------------------------------------
# LOAD TEST SET
# ------------------------------------------------

test_df = pd.read_csv(TEST_CSV)

print()
print(f"Test images: {len(test_df)}")
print()


# ------------------------------------------------
# CREATE DEGRADATIONS
# ------------------------------------------------

for degradation_name, degradation_function in DEGRADATIONS.items():

    print("-" * 70)
    print(f"Creating: {degradation_name}")

    output_dir = os.path.join(
        DEGRADATION_ROOT,
        degradation_name
    )

    os.makedirs(output_dir, exist_ok=True)

    records = []

    for idx, row in tqdm(
        test_df.iterrows(),
        total=len(test_df),
        desc=degradation_name
    ):

        image_path = row["path"]

        img = cv2.imread(
            image_path,
            cv2.IMREAD_GRAYSCALE
        )

        if img is None:
            print(f"WARNING: Could not read {image_path}")
            continue

        degraded = degradation_function(img)

        output_path = os.path.join(
            output_dir,
            f"{idx:05d}.png"
        )

        cv2.imwrite(
            output_path,
            degraded
        )

        records.append({
            "original_path": image_path,
            "degraded_path": output_path,
            "label": int(row["label"])
        })

    output_csv = os.path.join(
        DEGRADATION_ROOT,
        f"{degradation_name}.csv"
    )

    pd.DataFrame(records).to_csv(
        output_csv,
        index=False
    )

    print(f"Saved: {output_csv}")
    print(f"Images: {len(records)}")


print()
print("=" * 70)
print("✅ ALL DEGRADATION TEST SETS CREATED")
print("=" * 70)

print()
print("Location:")
print(DEGRADATION_ROOT)
```

```python
# ================================================================
# STEP 12 — CONFIDENCE vs IMAGE QUALITY AS ERROR PREDICTORS
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)

print("=" * 70)
print("CONFIDENCE vs IMAGE QUALITY — ERROR PREDICTION")
print("=" * 70)


# ------------------------------------------------
# LOAD QUALITY DATA
# ------------------------------------------------

conditions = [
    "clean",
    "blur",
    "noise",
    "low_resolution",
    "jpeg",
    "brightness_contrast"
]

quality_tables = {}

for condition in conditions:

    path = os.path.join(
        QUALITY_ROOT,
        f"{condition}_quality.csv"
    )

    df = pd.read_csv(path)

    quality_tables[condition] = df


# ------------------------------------------------
# IDENTIFY PROBABILITY / PREDICTION COLUMNS
# ------------------------------------------------

def get_probability_column(df):

    possible = [
        "probability_lung_opacity",
        "probability",
        "probability_lung_opacity "
    ]

    for col in possible:
        if col in df.columns:
            return col

    raise ValueError(
        f"Probability column not found. "
        f"Columns: {list(df.columns)}"
    )


def get_prediction_column(df):

    possible = [
        "predicted_label",
        "prediction"
    ]

    for col in possible:
        if col in df.columns:
            return col

    raise ValueError(
        f"Prediction column not found. "
        f"Columns: {list(df.columns)}"
    )


# ------------------------------------------------
# ANALYSIS
# ------------------------------------------------

analysis_rows = []

for condition, df in quality_tables.items():

    probability_column = get_probability_column(df)
    prediction_column = get_prediction_column(df)

    y_true = df["label"].values

    probabilities = df[
        probability_column
    ].values

    predictions = df[
        prediction_column
    ].values

    # ------------------------------------------------
    # MODEL CONFIDENCE
    # ------------------------------------------------

    confidence = np.maximum(
        probabilities,
        1 - probabilities
    )

    # Low confidence = more likely error
    uncertainty = 1 - confidence

    # ------------------------------------------------
    # ERROR
    # ------------------------------------------------

    error = (
        predictions != y_true
    ).astype(int)

    # ------------------------------------------------
    # QUALITY SIGNALS
    # ------------------------------------------------

    sharpness = df[
        "sharpness"
    ].values

    contrast = df[
        "contrast"
    ].values

    entropy = df[
        "entropy"
    ].values

    hf_variation = df[
        "high_frequency_variation"
    ].values

    # ------------------------------------------------
    # QUALITY-BASED ERROR SCORES
    # ------------------------------------------------
    #
    # For sharpness:
    # lower = worse quality
    #
    # For contrast:
    # lower = worse quality
    #
    # For entropy:
    # lower = potentially less information
    #
    # We use negative values so that
    # HIGHER score = more suspicious.
    # ------------------------------------------------

    quality_scores = {
        "uncertainty": uncertainty,

        "low_sharpness": -sharpness,

        "low_contrast": -contrast,

        "low_entropy": -entropy,

        "high_frequency": hf_variation
    }

    row = {
        "condition": condition,
        "error_rate": error.mean()
    }

    # ------------------------------------------------
    # AUROC FOR PREDICTING ERRORS
    # ------------------------------------------------

    for name, score in quality_scores.items():

        valid = np.isfinite(score)

        if (
            valid.sum() > 1
            and len(np.unique(error[valid])) == 2
        ):

            auroc = roc_auc_score(
                error[valid],
                score[valid]
            )

            ap = average_precision_score(
                error[valid],
                score[valid]
            )

        else:

            auroc = np.nan
            ap = np.nan

        row[f"{name}_AUROC"] = auroc
        row[f"{name}_AP"] = ap

    analysis_rows.append(row)


# ------------------------------------------------
# RESULTS
# ------------------------------------------------

error_prediction_results = pd.DataFrame(
    analysis_rows
)

print()
print("=" * 70)
print("ERROR PREDICTION RESULTS")
print("=" * 70)

print(
    error_prediction_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ------------------------------------------------
# SAVE
# ------------------------------------------------

error_prediction_path = os.path.join(
    QUALITY_ROOT,
    "error_prediction_results.csv"
)

error_prediction_results.to_csv(
    error_prediction_path,
    index=False
)


# ================================================================
# FIGURE — ERROR PREDICTION AUROC
# ================================================================

plot_columns = [
    "uncertainty_AUROC",
    "low_sharpness_AUROC",
    "low_contrast_AUROC",
    "low_entropy_AUROC",
    "high_frequency_AUROC"
]

labels = [
    "Confidence\n(uncertainty)",
    "Low\nsharpness",
    "Low\ncontrast",
    "Low\nentropy",
    "High-frequency\nvariation"
]

for condition in conditions:

    row = error_prediction_results[
        error_prediction_results["condition"] == condition
    ].iloc[0]

    values = [
        row[col]
        for col in plot_columns
    ]

    plt.figure(figsize=(9, 5))

    plt.bar(
        labels,
        values
    )

    plt.axhline(
        0.5,
        linestyle="--"
    )

    plt.ylabel(
        "AUROC for Predicting Prediction Error"
    )

    plt.title(
        f"Error Prediction Signals — {condition}"
    )

    plt.ylim(0, 1)

    plt.tight_layout()

    figure_path = os.path.join(
        FIGURE_ROOT,
        f"error_prediction_{condition}.png"
    )

    plt.savefig(
        figure_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


print()
print("=" * 70)
print("✅ ERROR PREDICTION ANALYSIS COMPLETE")
print("=" * 70)

print()
print("Results:")
print(error_prediction_path)

print()
print("Figures:")
print(FIGURE_ROOT)
```

```python
# ================================================================
# STEP 13 — SELECTIVE PREDICTION / RISK-COVERAGE ANALYSIS
# ================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


print("=" * 70)
print("SELECTIVE PREDICTION — RISK vs COVERAGE")
print("=" * 70)


# ------------------------------------------------
# LOAD QUALITY DATA
# ------------------------------------------------

conditions = [
    "clean",
    "blur",
    "noise",
    "low_resolution",
    "jpeg",
    "brightness_contrast"
]

quality_tables = {}

for condition in conditions:

    path = os.path.join(
        QUALITY_ROOT,
        f"{condition}_quality.csv"
    )

    quality_tables[condition] = pd.read_csv(path)


# ------------------------------------------------
# RISK-COVERAGE FUNCTION
# ------------------------------------------------

def risk_coverage(
    errors,
    scores
):
    """
    errors:
        1 = incorrect prediction
        0 = correct prediction

    scores:
        Higher = more trustworthy prediction.

    Returns:
        coverage
        risk
    """

    errors = np.asarray(errors)
    scores = np.asarray(scores)

    valid = (
        np.isfinite(errors)
        &
        np.isfinite(scores)
    )

    errors = errors[valid]
    scores = scores[valid]

    # Highest score = most trusted
    order = np.argsort(
        -scores
    )

    sorted_errors = errors[
        order
    ]

    cumulative_errors = np.cumsum(
        sorted_errors
    )

    n = len(sorted_errors)

    coverage = np.arange(
        1,
        n + 1
    ) / n

    risk = (
        cumulative_errors /
        np.arange(1, n + 1)
    )

    return coverage, risk


# ------------------------------------------------
# AURC
# ------------------------------------------------

def calculate_aurc(
    coverage,
    risk
):

    return np.trapezoid(
        risk,
        coverage
    )


# ------------------------------------------------
# ANALYSIS
# ------------------------------------------------

all_results = []

for condition in conditions:

    print()
    print("-" * 70)
    print(f"Condition: {condition}")

    df = quality_tables[
        condition
    ].copy()

    # ------------------------------------------------
    # PROBABILITY
    # ------------------------------------------------

    if "probability_lung_opacity" in df.columns:

        probability = df[
            "probability_lung_opacity"
        ].values

    else:

        probability = df[
            "probability"
        ].values


    # ------------------------------------------------
    # PREDICTION
    # ------------------------------------------------

    if "predicted_label" in df.columns:

        prediction = df[
            "predicted_label"
        ].values

    else:

        prediction = df[
            "prediction"
        ].values


    # ------------------------------------------------
    # TRUE LABEL
    # ------------------------------------------------

    true_label = df[
        "label"
    ].values


    # ------------------------------------------------
    # ERROR
    # ------------------------------------------------

    errors = (
        prediction != true_label
    ).astype(int)


    # ------------------------------------------------
    # CONFIDENCE
    # ------------------------------------------------

    confidence = np.maximum(
        probability,
        1 - probability
    )


    # ------------------------------------------------
    # QUALITY SIGNAL
    # ------------------------------------------------
    #
    # Higher = better quality
    #
    # Sharpness is the main quality
    # signal because blur and
    # low-resolution strongly affect it.
    # ------------------------------------------------

    sharpness = df[
        "sharpness"
    ].values

    # Normalize using percentile
    # to reduce extreme noise influence

    q_low = np.percentile(
        sharpness,
        1
    )

    q_high = np.percentile(
        sharpness,
        99
    )

    quality_score = np.clip(
        (sharpness - q_low) /
        (q_high - q_low + 1e-8),
        0,
        1
    )


    # ------------------------------------------------
    # COMBINED SCORE
    # ------------------------------------------------
    #
    # Geometric mean:
    #
    # high confidence AND
    # high quality are required.
    # ------------------------------------------------

    combined_score = np.sqrt(
        confidence *
        quality_score
    )


    # ------------------------------------------------
    # RISK-COVERAGE
    # ------------------------------------------------

    scores = {
        "confidence": confidence,

        "quality": quality_score,

        "combined": combined_score
    }


    condition_results = {}

    for method, score in scores.items():

        coverage, risk = risk_coverage(
            errors,
            score
        )

        aurc = calculate_aurc(
            coverage,
            risk
        )

        condition_results[
            method
        ] = (
            coverage,
            risk,
            aurc
        )

        # ------------------------------------------------
        # RISKS AT FIXED COVERAGE
        # ------------------------------------------------

        target_coverages = [
            0.50,
            0.60,
            0.70,
            0.80,
            0.90,
            1.00
        ]

        for target in target_coverages:

            index = np.argmin(
                np.abs(
                    coverage -
                    target
                )
            )

            all_results.append({

                "condition":
                    condition,

                "method":
                    method,

                "coverage":
                    coverage[index],

                "risk":
                    risk[index],

                "aurc":
                    aurc
            })


        print(
            f"{method:10s} "
            f"AURC = {aurc:.4f}"
        )


    # ------------------------------------------------
    # PLOT
    # ------------------------------------------------

    plt.figure(
        figsize=(8, 6)
    )

    for method in scores:

        coverage, risk, aurc = (
            condition_results[
                method
            ]
        )

        plt.plot(
            coverage,
            risk,
            label=f"{method} (AURC={aurc:.4f})"
        )

    plt.xlabel(
        "Coverage"
    )

    plt.ylabel(
        "Selective Risk"
    )

    plt.title(
        f"Risk-Coverage Curve — {condition}"
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    figure_path = os.path.join(
        FIGURE_ROOT,
        f"risk_coverage_{condition}.png"
    )

    plt.savefig(
        figure_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()


# ================================================================
# SAVE RESULTS
# ================================================================

risk_coverage_results = pd.DataFrame(
    all_results
)

results_path = os.path.join(
    QUALITY_ROOT,
    "risk_coverage_results.csv"
)

risk_coverage_results.to_csv(
    results_path,
    index=False
)


# ================================================================
# SUMMARY TABLE
# ================================================================

aurc_summary = (
    risk_coverage_results
    .groupby(
        ["condition", "method"]
    )["aurc"]
    .first()
    .reset_index()
)

print()
print("=" * 70)
print("AURC SUMMARY")
print("=" * 70)

print(
    aurc_summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.5f}"
    )
)

print()
print("=" * 70)
print("✅ RISK-COVERAGE ANALYSIS COMPLETE")
print("=" * 70)

print()
print("Results:")
print(results_path)

print()
print("Figures:")
print(FIGURE_ROOT)
```

```python
# ================================================================
# STEP 14 — GENERATE VALIDATION PREDICTIONS + QUALITY FEATURES
# ================================================================

import os
import cv2
import torch
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm

print("=" * 70)
print("GENERATING VALIDATION RELIABILITY FEATURES")
print("=" * 70)

# ------------------------------------------------
# PATHS
# ------------------------------------------------

VALIDATION_CSV = os.path.join(
    RESEARCH_DIR,
    "validation.csv"
)

VALIDATION_OUTPUT = os.path.join(
    QUALITY_ROOT,
    "validation_reliability_features.csv"
)

validation_df = pd.read_csv(
    VALIDATION_CSV
)

print(f"Validation images: {len(validation_df)}")


# ------------------------------------------------
# LOAD BEST MODEL
# ------------------------------------------------

checkpoint = torch.load(
    BEST_MODEL_PATH,
    map_location=device,
    weights_only=False
)

model.load_state_dict(
    checkpoint["model_state_dict"]
)

model.eval()


# ------------------------------------------------
# QUALITY FUNCTIONS
# ------------------------------------------------

def calculate_sharpness(img):

    return float(
        cv2.Laplacian(
            img,
            cv2.CV_64F
        ).var()
    )


def calculate_contrast(img):

    return float(
        np.std(img)
    )


def calculate_entropy(img):

    histogram = cv2.calcHist(
        [img],
        [0],
        None,
        [256],
        [0, 256]
    ).flatten()

    histogram = histogram[
        histogram > 0
    ]

    probabilities = (
        histogram /
        histogram.sum()
    )

    return float(
        -np.sum(
            probabilities *
            np.log2(probabilities)
        )
    )


def calculate_hf_variation(img):

    laplacian = cv2.Laplacian(
        img,
        cv2.CV_64F
    )

    return float(
        np.std(laplacian)
    )


# ------------------------------------------------
# TRANSFORM
# ------------------------------------------------

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ------------------------------------------------
# ARRAYS
# ------------------------------------------------

probabilities = []
predictions = []

sharpness_values = []
contrast_values = []
entropy_values = []
hf_values = []


# ------------------------------------------------
# PROCESS VALIDATION SET
# ------------------------------------------------

for image_path in tqdm(
    validation_df["path"],
    total=len(validation_df),
    desc="Validation"
):

    # --------------------------------------------
    # IMAGE
    # --------------------------------------------

    img = cv2.imread(
        image_path,
        cv2.IMREAD_GRAYSCALE
    )

    if img is None:

        probabilities.append(np.nan)
        predictions.append(np.nan)

        sharpness_values.append(np.nan)
        contrast_values.append(np.nan)
        entropy_values.append(np.nan)
        hf_values.append(np.nan)

        continue


    # --------------------------------------------
    # QUALITY
    # --------------------------------------------

    sharpness_values.append(
        calculate_sharpness(img)
    )

    contrast_values.append(
        calculate_contrast(img)
    )

    entropy_values.append(
        calculate_entropy(img)
    )

    hf_values.append(
        calculate_hf_variation(img)
    )


    # --------------------------------------------
    # MODEL PREDICTION
    # --------------------------------------------

    image = Image.fromarray(
        img
    ).convert("RGB")

    tensor = eval_transform(
        image
    ).unsqueeze(0).to(device)


    with torch.no_grad():

        output = model(
            tensor
        )

        probability = torch.softmax(
            output,
            dim=1
        )[0, 1].item()


    prediction = int(
        probability >= 0.5
    )

    probabilities.append(
        probability
    )

    predictions.append(
        prediction
    )


# ------------------------------------------------
# BUILD TABLE
# ------------------------------------------------

validation_features = validation_df.copy()

validation_features[
    "probability"
] = probabilities

validation_features[
    "prediction"
] = predictions

validation_features[
    "confidence"
] = np.maximum(
    validation_features["probability"],
    1 - validation_features["probability"]
)

validation_features[
    "error"
] = (
    validation_features["prediction"]
    != validation_features["label"]
).astype(int)

validation_features[
    "sharpness"
] = sharpness_values

validation_features[
    "contrast"
] = contrast_values

validation_features[
    "entropy"
] = entropy_values

validation_features[
    "high_frequency_variation"
] = hf_values


# ------------------------------------------------
# SAVE
# ------------------------------------------------

validation_features.to_csv(
    VALIDATION_OUTPUT,
    index=False
)


# ------------------------------------------------
# SUMMARY
# ------------------------------------------------

print()
print("=" * 70)
print("VALIDATION FEATURE SUMMARY")
print("=" * 70)

print(
    f"Samples: "
    f"{len(validation_features)}"
)

print(
    f"Errors: "
    f"{validation_features['error'].sum()}"
)

print(
    f"Error rate: "
    f"{validation_features['error'].mean():.4f}"
)

print()
print("Saved:")
print(VALIDATION_OUTPUT)

print()
print("=" * 70)
print("✅ VALIDATION FEATURES READY")
print("=" * 70)
```

```python
# ================================================================
# STEP 15 — TRAIN RELIABILITY MODEL
# ================================================================

import os
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    classification_report
)

print("=" * 70)
print("TRAINING RELIABILITY MODEL")
print("=" * 70)

VALIDATION_OUTPUT = os.path.join(
    QUALITY_ROOT,
    "validation_reliability_features.csv"
)

RELIABILITY_MODEL_PATH = os.path.join(
    QUALITY_ROOT,
    "reliability_model.pkl"
)

df = pd.read_csv(VALIDATION_OUTPUT)

# ------------------------------------------------
# Features
# ------------------------------------------------

feature_columns = [
    "confidence",
    "sharpness",
    "contrast",
    "entropy",
    "high_frequency_variation"
]

X = df[feature_columns].copy()
y = df["error"].astype(int)

print(f"Samples : {len(df)}")
print(f"Errors  : {y.sum()}")
print(f"Correct : {(y == 0).sum()}")

print()
print("Features:")
for feature in feature_columns:
    print(f"  - {feature}")

# ------------------------------------------------
# Split validation data
# 70% reliability training
# 30% reliability holdout
# ------------------------------------------------

X_train, X_holdout, y_train, y_holdout = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print()
print(f"Reliability training samples : {len(X_train)}")
print(f"Reliability holdout samples  : {len(X_holdout)}")

print()
print("Training error rate : "
      f"{y_train.mean():.4f}")

print("Holdout error rate  : "
      f"{y_holdout.mean():.4f}")

# ------------------------------------------------
# Logistic regression pipeline
# ------------------------------------------------

reliability_model = Pipeline([
    (
        "scaler",
        StandardScaler()
    ),
    (
        "classifier",
        LogisticRegression(
            class_weight="balanced",
            max_iter=2000,
            random_state=42
        )
    )
])

# ------------------------------------------------
# Train
# ------------------------------------------------

print()
print("Training reliability model...")

reliability_model.fit(
    X_train,
    y_train
)

# ------------------------------------------------
# Evaluate ONLY on reliability holdout
# ------------------------------------------------

holdout_probability = reliability_model.predict_proba(
    X_holdout
)[:, 1]

holdout_auc = roc_auc_score(
    y_holdout,
    holdout_probability
)

holdout_ap = average_precision_score(
    y_holdout,
    holdout_probability
)

print()
print("=" * 70)
print("RELIABILITY MODEL HOLDOUT PERFORMANCE")
print("=" * 70)

print(
    f"AUROC : {holdout_auc:.4f}"
)

print(
    f"AP    : {holdout_ap:.4f}"
)

# ------------------------------------------------
# Feature coefficients
# ------------------------------------------------

classifier = reliability_model.named_steps[
    "classifier"
]

print()
print("Feature coefficients:")

for feature, coefficient in zip(
    feature_columns,
    classifier.coef_[0]
):
    print(
        f"{feature:30s}: {coefficient:+.4f}"
    )

# ------------------------------------------------
# Save model
# ------------------------------------------------

joblib.dump(
    reliability_model,
    RELIABILITY_MODEL_PATH
)

print()
print("Saved:")
print(RELIABILITY_MODEL_PATH)

print()
print("=" * 70)
print("✅ RELIABILITY MODEL READY")
print("=" * 70)
```

```python
# ================================================================
# STEP 16 — FROZEN RELIABILITY MODEL ON TEST CONDITIONS
# ================================================================

import os
import joblib
import numpy as np
import pandas as pd
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score
)

print("=" * 70)
print("FROZEN RELIABILITY MODEL — TEST EVALUATION")
print("=" * 70)

# ------------------------------------------------
# Load frozen reliability model
# ------------------------------------------------

RELIABILITY_MODEL_PATH = os.path.join(
    QUALITY_ROOT,
    "reliability_model.pkl"
)

reliability_model = joblib.load(
    RELIABILITY_MODEL_PATH
)

# ------------------------------------------------
# Test condition files
# ------------------------------------------------

condition_files = {
    "clean": os.path.join(
        QUALITY_ROOT,
        "clean_quality.csv"
    ),
    "blur": os.path.join(
        QUALITY_ROOT,
        "blur_quality.csv"
    ),
    "noise": os.path.join(
        QUALITY_ROOT,
        "noise_quality.csv"
    ),
    "low_resolution": os.path.join(
        QUALITY_ROOT,
        "low_resolution_quality.csv"
    ),
    "jpeg": os.path.join(
        QUALITY_ROOT,
        "jpeg_quality.csv"
    ),
    "brightness_contrast": os.path.join(
        QUALITY_ROOT,
        "brightness_contrast_quality.csv"
    )
}

# ------------------------------------------------
# Prediction files
# ------------------------------------------------

prediction_files = {
    "clean": os.path.join(
        RESEARCH_DIR,
        "clean_test_predictions.csv"
    ),
    "blur": os.path.join(
        RESEARCH_DIR,
        "degradation_test_sets",
        "blur.csv"
    ),
    "noise": os.path.join(
        RESEARCH_DIR,
        "degradation_test_sets",
        "noise.csv"
    ),
    "low_resolution": os.path.join(
        RESEARCH_DIR,
        "degradation_test_sets",
        "low_resolution.csv"
    ),
    "jpeg": os.path.join(
        RESEARCH_DIR,
        "degradation_test_sets",
        "jpeg.csv"
    ),
    "brightness_contrast": os.path.join(
        RESEARCH_DIR,
        "degradation_test_sets",
        "brightness_contrast.csv"
    )
}

feature_columns = [
    "confidence",
    "sharpness",
    "contrast",
    "entropy",
    "high_frequency_variation"
]

results = []

# ------------------------------------------------
# Evaluate each condition
# ------------------------------------------------

for condition in condition_files:

    print()
    print("-" * 70)
    print(f"CONDITION: {condition.upper()}")
    print("-" * 70)

    quality_df = pd.read_csv(
        condition_files[condition]
    )

    prediction_df = pd.read_csv(
        prediction_files[condition]
    )

    # ------------------------------------------------
    # Merge by path
    # ------------------------------------------------

    df = prediction_df.merge(
        quality_df[
            [
                "path",
                "sharpness",
                "contrast",
                "entropy",
                "high_frequency_variation"
            ]
        ],
        on="path",
        how="inner"
    )

    # ------------------------------------------------
    # Safety check
    # ------------------------------------------------

    print(
        f"Samples after merge: {len(df)}"
    )

    if len(df) == 0:
        print("WARNING: no matching samples")
        continue

    # ------------------------------------------------
    # Recalculate confidence
    # ------------------------------------------------

    df["confidence"] = np.maximum(
        df["probability"],
        1 - df["probability"]
    )

    # ------------------------------------------------
    # Error target
    # ------------------------------------------------

    df["error"] = (
        df["prediction"] != df["label"]
    ).astype(int)

    # ------------------------------------------------
    # 1. CONFIDENCE ONLY
    # ------------------------------------------------

    confidence_score = df[
        "confidence"
    ].values

    confidence_uncertainty = (
        1 - confidence_score
    )

    confidence_auc = roc_auc_score(
        df["error"],
        confidence_uncertainty
    )

    confidence_ap = average_precision_score(
        df["error"],
        confidence_uncertainty
    )

    # ------------------------------------------------
    # 2. QUALITY ONLY
    #
    # Reliability model without confidence
    # is NOT retrained here.
    #
    # Instead use normalized sharpness as
    # a simple quality baseline.
    # ------------------------------------------------

    sharpness = df[
        "sharpness"
    ].values

    q1 = np.percentile(
        sharpness,
        1
    )

    q99 = np.percentile(
        sharpness,
        99
    )

    quality_score = np.clip(
        (sharpness - q1) /
        (q99 - q1 + 1e-8),
        0,
        1
    )

    quality_uncertainty = (
        1 - quality_score
    )

    quality_auc = roc_auc_score(
        df["error"],
        quality_uncertainty
    )

    quality_ap = average_precision_score(
        df["error"],
        quality_uncertainty
    )

    # ------------------------------------------------
    # 3. CONFIDENCE + QUALITY
    # ------------------------------------------------

    X = df[
        feature_columns
    ]

    reliability_probability = (
        reliability_model
        .predict_proba(X)[:, 1]
    )

    combined_auc = roc_auc_score(
        df["error"],
        reliability_probability
    )

    combined_ap = average_precision_score(
        df["error"],
        reliability_probability
    )

    # ------------------------------------------------
    # Store results
    # ------------------------------------------------

    results.append({
        "condition": condition,

        "error_rate":
            df["error"].mean(),

        "confidence_AUROC":
            confidence_auc,

        "confidence_AP":
            confidence_ap,

        "quality_AUROC":
            quality_auc,

        "quality_AP":
            quality_ap,

        "reliability_AUROC":
            combined_auc,

        "reliability_AP":
            combined_ap
    })

    print()
    print(
        f"Error rate          : "
        f"{df['error'].mean():.4f}"
    )

    print(
        f"Confidence AUROC    : "
        f"{confidence_auc:.4f}"
    )

    print(
        f"Confidence AP       : "
        f"{confidence_ap:.4f}"
    )

    print(
        f"Quality AUROC       : "
        f"{quality_auc:.4f}"
    )

    print(
        f"Quality AP          : "
        f"{quality_ap:.4f}"
    )

    print(
        f"Reliability AUROC   : "
        f"{combined_auc:.4f}"
    )

    print(
        f"Reliability AP      : "
        f"{combined_ap:.4f}"
    )

# ------------------------------------------------
# Final table
# ------------------------------------------------

results_df = pd.DataFrame(
    results
)

RESULT_PATH = os.path.join(
    QUALITY_ROOT,
    "frozen_reliability_test_results.csv"
)

results_df.to_csv(
    RESULT_PATH,
    index=False
)

print()
print("=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print(
    results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

print()
print("Saved:")
print(RESULT_PATH)

print()
print("=" * 70)
print("✅ FROZEN RELIABILITY EVALUATION COMPLETE")
print("=" * 70)
```

```python
# =====================================================================
# INCREMENTAL VALUE TEST
# Does image quality add information beyond model confidence?
# =====================================================================

import os
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, average_precision_score

print("=" * 75)
print("INCREMENTAL VALUE TEST")
print("Confidence-only vs Confidence + Image Quality")
print("=" * 75)

# ---------------------------------------------------------------------
# PATH
# ---------------------------------------------------------------------

BASE_DIR = r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset\research_dataset"

FEATURE_FILE = os.path.join(
    BASE_DIR,
    "quality_analysis",
    "validation_reliability_features.csv"
)

OUTPUT_FILE = os.path.join(
    BASE_DIR,
    "quality_analysis",
    "incremental_value_validation_results.csv"
)

# ---------------------------------------------------------------------
# LOAD VALIDATION FEATURES
# ---------------------------------------------------------------------

df = pd.read_csv(FEATURE_FILE)

print("\nValidation samples:", len(df))
print("Columns:")
print(df.columns.tolist())

# ---------------------------------------------------------------------
# TARGET
# ---------------------------------------------------------------------

# Error = 1 if classifier prediction was wrong
y = df["error"].astype(int).values

print("\nValidation errors:", y.sum())
print("Validation correct:", (y == 0).sum())
print("Error rate:", round(y.mean(), 4))

# ---------------------------------------------------------------------
# FEATURES
# ---------------------------------------------------------------------

confidence_features = [
    "confidence"
]

quality_features = [
    "sharpness",
    "contrast",
    "entropy",
    "high_frequency_variation"
]

full_features = confidence_features + quality_features

# Check that everything exists
missing = [c for c in full_features if c not in df.columns]

if missing:
    raise KeyError(f"Missing required columns: {missing}")

X_conf = df[confidence_features].values
X_full = df[full_features].values

# ---------------------------------------------------------------------
# 5-FOLD STRATIFIED CROSS-VALIDATION
# ---------------------------------------------------------------------

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

confidence_auc = []
confidence_ap = []

full_auc = []
full_ap = []

fold_results = []

print("\n" + "-" * 75)
print("RUNNING 5-FOLD CROSS-VALIDATION")
print("-" * 75)

for fold, (train_idx, val_idx) in enumerate(skf.split(X_full, y), start=1):

    # -------------------------------------------------------------
    # MODEL A: CONFIDENCE ONLY
    # -------------------------------------------------------------

    model_conf = Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        ))
    ])

    model_conf.fit(
        X_conf[train_idx],
        y[train_idx]
    )

    pred_conf = model_conf.predict_proba(
        X_conf[val_idx]
    )[:, 1]

    auc_conf = roc_auc_score(
        y[val_idx],
        pred_conf
    )

    ap_conf = average_precision_score(
        y[val_idx],
        pred_conf
    )

    # -------------------------------------------------------------
    # MODEL B: CONFIDENCE + QUALITY
    # -------------------------------------------------------------

    model_full = Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        ))
    ])

    model_full.fit(
        X_full[train_idx],
        y[train_idx]
    )

    pred_full = model_full.predict_proba(
        X_full[val_idx]
    )[:, 1]

    auc_full = roc_auc_score(
        y[val_idx],
        pred_full
    )

    ap_full_fold = average_precision_score(
        y[val_idx],
        pred_full
    )

    # -------------------------------------------------------------
    # STORE
    # -------------------------------------------------------------

    confidence_auc.append(auc_conf)
    confidence_ap.append(ap_conf)

    full_auc.append(auc_full)
    full_ap.append(ap_full_fold)

    delta_auc = auc_full - auc_conf
    delta_ap = ap_full_fold - ap_conf

    fold_results.append({
        "fold": fold,
        "confidence_AUROC": auc_conf,
        "full_AUROC": auc_full,
        "delta_AUROC": delta_auc,
        "confidence_AP": ap_conf,
        "full_AP": ap_full_fold,
        "delta_AP": delta_ap
    })

    print(
        f"Fold {fold}/5 | "
        f"Confidence AUROC: {auc_conf:.4f} | "
        f"Full AUROC: {auc_full:.4f} | "
        f"Δ AUROC: {delta_auc:+.4f}"
    )

# ---------------------------------------------------------------------
# RESULTS
# ---------------------------------------------------------------------

results_df = pd.DataFrame(fold_results)

print("\n" + "=" * 75)
print("5-FOLD RESULTS")
print("=" * 75)

print(results_df.to_string(index=False))

# ---------------------------------------------------------------------
# MEANS + STANDARD DEVIATION
# ---------------------------------------------------------------------

conf_auc_mean = np.mean(confidence_auc)
conf_auc_std = np.std(confidence_auc, ddof=1)

full_auc_mean = np.mean(full_auc)
full_auc_std = np.std(full_auc, ddof=1)

conf_ap_mean = np.mean(confidence_ap)
conf_ap_std = np.std(confidence_ap, ddof=1)

full_ap_mean = np.mean(full_ap)
full_ap_std = np.std(full_ap, ddof=1)

mean_delta_auc = full_auc_mean - conf_auc_mean
mean_delta_ap = full_ap_mean - conf_ap_mean

print("\n" + "=" * 75)
print("SUMMARY")
print("=" * 75)

print(
    f"\nConfidence-only AUROC : "
    f"{conf_auc_mean:.4f} ± {conf_auc_std:.4f}"
)

print(
    f"Confidence + Quality AUROC : "
    f"{full_auc_mean:.4f} ± {full_auc_std:.4f}"
)

print(
    f"Mean AUROC improvement : "
    f"{mean_delta_auc:+.4f}"
)

print(
    f"\nConfidence-only AP : "
    f"{conf_ap_mean:.4f} ± {conf_ap_std:.4f}"
)

print(
    f"Confidence + Quality AP : "
    f"{full_ap_mean:.4f} ± {full_ap_std:.4f}"
)

print(
    f"Mean AP improvement : "
    f"{mean_delta_ap:+.4f}"
)

# ---------------------------------------------------------------------
# SAVE
# ---------------------------------------------------------------------

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\nSaved:")
print(OUTPUT_FILE)

print("\n" + "=" * 75)
print("INCREMENTAL VALUE TEST COMPLETE")
print("=" * 75)
```

```python
# =====================================================================
# FINAL SELECTIVE PREDICTION / ABSTENTION EVALUATION
# Confidence vs Quality vs Reliability
# =====================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import auc

print("=" * 75)
print("SELECTIVE PREDICTION / ABSTENTION EVALUATION")
print("=" * 75)

# ---------------------------------------------------------------------
# PATHS
# ---------------------------------------------------------------------

BASE_DIR = r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset\research_dataset"

QUALITY_DIR = os.path.join(BASE_DIR, "quality_analysis")

PRED_DIR = os.path.join(
    BASE_DIR,
    "degradation_test_predictions"
)

OUTPUT_CSV = os.path.join(
    QUALITY_DIR,
    "final_selective_prediction_results.csv"
)

FIG_DIR = os.path.join(
    QUALITY_DIR,
    "final_risk_coverage_figures"
)

os.makedirs(FIG_DIR, exist_ok=True)

# ---------------------------------------------------------------------
# CONDITIONS
# ---------------------------------------------------------------------

conditions = {
    "clean": (
        os.path.join(BASE_DIR, "clean_test_predictions.csv"),
        os.path.join(QUALITY_DIR, "clean_quality.csv")
    ),
    "blur": (
        os.path.join(PRED_DIR, "blur_predictions.csv"),
        os.path.join(QUALITY_DIR, "blur_quality.csv")
    ),
    "noise": (
        os.path.join(PRED_DIR, "noise_predictions.csv"),
        os.path.join(QUALITY_DIR, "noise_quality.csv")
    ),
    "low_resolution": (
        os.path.join(PRED_DIR, "low_resolution_predictions.csv"),
        os.path.join(QUALITY_DIR, "low_resolution_quality.csv")
    ),
    "jpeg": (
        os.path.join(PRED_DIR, "jpeg_predictions.csv"),
        os.path.join(QUALITY_DIR, "jpeg_quality.csv")
    ),
    "brightness_contrast": (
        os.path.join(PRED_DIR, "brightness_contrast_predictions.csv"),
        os.path.join(QUALITY_DIR, "brightness_contrast_quality.csv")
    )
}

# ---------------------------------------------------------------------
# LOAD FROZEN RELIABILITY MODEL
# ---------------------------------------------------------------------

import joblib

RELIABILITY_MODEL_PATH = os.path.join(
    QUALITY_DIR,
    "reliability_model.pkl"
)

reliability_model = joblib.load(RELIABILITY_MODEL_PATH)

print("\nFrozen reliability model loaded.")

# ---------------------------------------------------------------------
# HELPER: RISK-COVERAGE
# ---------------------------------------------------------------------

def calculate_risk_coverage(trust_score, errors):

    trust_score = np.asarray(trust_score)
    errors = np.asarray(errors)

    # Highest trust first
    order = np.argsort(-trust_score)

    sorted_errors = errors[order]

    cumulative_errors = np.cumsum(sorted_errors)

    n = len(errors)

    coverage = np.arange(1, n + 1) / n

    risk = cumulative_errors / np.arange(1, n + 1)

    return coverage, risk


def calculate_aurc(coverage, risk):

    # Add origin (0,0)
    x = np.concatenate([[0], coverage])
    y = np.concatenate([[0], risk])

    return np.trapezoid(y, x)


def risk_at_coverage(coverage, risk, target):

    idx = np.argmin(
        np.abs(coverage - target)
    )

    return risk[idx]


# ---------------------------------------------------------------------
# MAIN LOOP
# ---------------------------------------------------------------------

all_results = []
curve_data = {}

for condition, (pred_file, quality_file) in conditions.items():

    print("\n" + "-" * 75)
    print(f"CONDITION: {condition.upper()}")
    print("-" * 75)

    pred_df = pd.read_csv(pred_file)
    quality_df = pd.read_csv(quality_file)

    print("Prediction samples:", len(pred_df))
    print("Quality samples   :", len(quality_df))

    # -------------------------------------------------------------
    # ALIGN DATA
    # -------------------------------------------------------------

    if (
        "path" in pred_df.columns
        and "path" in quality_df.columns
    ):

        merged = pred_df.merge(
            quality_df[
                [
                    "path",
                    "sharpness",
                    "contrast",
                    "entropy",
                    "high_frequency_variation"
                ]
            ],
            on="path",
            how="inner"
        )

        alignment = "path"

    else:

        if len(pred_df) != len(quality_df):
            raise ValueError(
                f"{condition}: prediction and quality lengths differ."
            )

        merged = pred_df.copy()

        for col in [
            "sharpness",
            "contrast",
            "entropy",
            "high_frequency_variation"
        ]:
            merged[col] = quality_df[col].values

        alignment = "row_order"

    print("Alignment:", alignment)
    print("Samples after alignment:", len(merged))

    # -------------------------------------------------------------
    # TARGET / ERROR
    # -------------------------------------------------------------

    if "true_label" in merged.columns:
        true_label = merged["true_label"].values
    else:
        true_label = merged["label"].values

    if "predicted_label" in merged.columns:
        predicted_label = merged["predicted_label"].values
    else:
        predicted_label = merged["prediction"].values

    errors = (
        predicted_label != true_label
    ).astype(int)

    # -------------------------------------------------------------
    # CONFIDENCE
    # -------------------------------------------------------------

    if "probability_lung_opacity" in merged.columns:

        p = merged[
            "probability_lung_opacity"
        ].values

    elif "probability" in merged.columns:

        p = merged[
            "probability"
        ].values

    else:
        raise KeyError(
            f"{condition}: probability column not found."
        )

    confidence = np.maximum(
        p,
        1 - p
    )

    # -------------------------------------------------------------
    # QUALITY SCORE
    # -------------------------------------------------------------

    sharpness = merged[
        "sharpness"
    ].values.astype(float)

    # Robust normalization using 1st / 99th percentile
    q_low = np.percentile(
        sharpness,
        1
    )

    q_high = np.percentile(
        sharpness,
        99
    )

    quality_score = (
        (sharpness - q_low)
        /
        (q_high - q_low + 1e-8)
    )

    quality_score = np.clip(
        quality_score,
        0,
        1
    )

    # -------------------------------------------------------------
    # FROZEN RELIABILITY MODEL
    # -------------------------------------------------------------

    reliability_X = pd.DataFrame({
        "confidence": confidence,
        "sharpness": merged["sharpness"].values,
        "contrast": merged["contrast"].values,
        "entropy": merged["entropy"].values,
        "high_frequency_variation":
            merged["high_frequency_variation"].values
    })

    predicted_error_probability = (
        reliability_model.predict_proba(
            reliability_X
        )[:, 1]
    )

    # Higher trust = lower predicted error
    reliability_trust = (
        1 - predicted_error_probability
    )

    # -------------------------------------------------------------
    # RISK-COVERAGE
    # -------------------------------------------------------------

    coverage_conf, risk_conf = (
        calculate_risk_coverage(
            confidence,
            errors
        )
    )

    coverage_quality, risk_quality = (
        calculate_risk_coverage(
            quality_score,
            errors
        )
    )

    coverage_reliability, risk_reliability = (
        calculate_risk_coverage(
            reliability_trust,
            errors
        )
    )

    # -------------------------------------------------------------
    # AURC
    # -------------------------------------------------------------

    aurc_conf = calculate_aurc(
        coverage_conf,
        risk_conf
    )

    aurc_quality = calculate_aurc(
        coverage_quality,
        risk_quality
    )

    aurc_reliability = calculate_aurc(
        coverage_reliability,
        risk_reliability
    )

    # -------------------------------------------------------------
    # PRINT
    # -------------------------------------------------------------

    print(f"\nAURC:")
    print(f"  Confidence   : {aurc_conf:.6f}")
    print(f"  Quality      : {aurc_quality:.6f}")
    print(f"  Reliability  : {aurc_reliability:.6f}")

    print("\nRisk at selected coverage:")

    for target in [0.50, 0.75, 0.90, 0.95]:

        r_conf = risk_at_coverage(
            coverage_conf,
            risk_conf,
            target
        )

        r_quality = risk_at_coverage(
            coverage_quality,
            risk_quality,
            target
        )

        r_reliability = risk_at_coverage(
            coverage_reliability,
            risk_reliability,
            target
        )

        print(
            f"\nCoverage {target:.0%}"
        )

        print(
            f"  Confidence  : {r_conf:.4f}"
        )

        print(
            f"  Quality     : {r_quality:.4f}"
        )

        print(
            f"  Reliability : {r_reliability:.4f}"
        )

        all_results.append({
            "condition": condition,
            "coverage": target,
            "confidence_risk": r_conf,
            "quality_risk": r_quality,
            "reliability_risk": r_reliability
        })

    # -------------------------------------------------------------
    # SAVE CURVES
    # -------------------------------------------------------------

    curve_data[condition] = {
        "coverage_confidence": coverage_conf,
        "risk_confidence": risk_conf,

        "coverage_quality": coverage_quality,
        "risk_quality": risk_quality,

        "coverage_reliability": coverage_reliability,
        "risk_reliability": risk_reliability
    }

    # -------------------------------------------------------------
    # PLOT
    # -------------------------------------------------------------

    plt.figure(figsize=(8, 6))

    plt.plot(
        coverage_conf,
        risk_conf,
        label="Confidence"
    )

    plt.plot(
        coverage_quality,
        risk_quality,
        label="Quality"
    )

    plt.plot(
        coverage_reliability,
        risk_reliability,
        label="Reliability"
    )

    plt.xlabel("Coverage")
    plt.ylabel("Selective Risk")
    plt.title(
        f"Risk-Coverage Curve — {condition}"
    )

    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()

    figure_path = os.path.join(
        FIG_DIR,
        f"risk_coverage_{condition}.png"
    )

    plt.savefig(
        figure_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()

# ---------------------------------------------------------------------
# SAVE FINAL TABLE
# ---------------------------------------------------------------------

results_df = pd.DataFrame(
    all_results
)

results_df.to_csv(
    OUTPUT_CSV,
    index=False
)

print("\n" + "=" * 75)
print("FINAL SELECTIVE PREDICTION RESULTS")
print("=" * 75)

print(
    results_df.to_string(index=False)
)

print("\nSaved:")
print(OUTPUT_CSV)

print("\nFigures:")
print(FIG_DIR)

print("\n" + "=" * 75)
print("SELECTIVE PREDICTION EVALUATION COMPLETE")
print("=" * 75)
```

```python
# =====================================================================
# CALIBRATION ANALYSIS — FROZEN DENSENET121
# Clean + all degradation conditions
# =====================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import brier_score_loss
from sklearn.calibration import calibration_curve

print("=" * 75)
print("CALIBRATION ANALYSIS")
print("=" * 75)

BASE_DIR = r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset\research_dataset"

PRED_DIR = os.path.join(
    BASE_DIR,
    "degradation_test_predictions"
)

OUTPUT_DIR = os.path.join(
    BASE_DIR,
    "quality_analysis",
    "calibration"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

conditions = {
    "clean": os.path.join(
        BASE_DIR,
        "clean_test_predictions.csv"
    ),
    "blur": os.path.join(
        PRED_DIR,
        "blur_predictions.csv"
    ),
    "noise": os.path.join(
        PRED_DIR,
        "noise_predictions.csv"
    ),
    "low_resolution": os.path.join(
        PRED_DIR,
        "low_resolution_predictions.csv"
    ),
    "jpeg": os.path.join(
        PRED_DIR,
        "jpeg_predictions.csv"
    ),
    "brightness_contrast": os.path.join(
        PRED_DIR,
        "brightness_contrast_predictions.csv"
    )
}


# ---------------------------------------------------------------------
# EXPECTED CALIBRATION ERROR
# ---------------------------------------------------------------------

def expected_calibration_error(
    y_true,
    probabilities,
    n_bins=10
):

    y_true = np.asarray(y_true)
    probabilities = np.asarray(probabilities)

    bins = np.linspace(0, 1, n_bins + 1)

    ece = 0.0

    bin_rows = []

    for i in range(n_bins):

        if i == n_bins - 1:
            mask = (
                (probabilities >= bins[i]) &
                (probabilities <= bins[i + 1])
            )
        else:
            mask = (
                (probabilities >= bins[i]) &
                (probabilities < bins[i + 1])
            )

        count = mask.sum()

        if count == 0:
            continue

        confidence = probabilities[mask].mean()
        accuracy = y_true[mask].mean()

        weight = count / len(y_true)

        ece += weight * abs(
            accuracy - confidence
        )

        bin_rows.append({
            "bin": i,
            "count": count,
            "confidence": confidence,
            "accuracy": accuracy
        })

    return ece, pd.DataFrame(bin_rows)


# ---------------------------------------------------------------------
# PROCESS CONDITIONS
# ---------------------------------------------------------------------

summary = []

for condition, pred_file in conditions.items():

    print("\n" + "-" * 75)
    print(f"CONDITION: {condition.upper()}")
    print("-" * 75)

    df = pd.read_csv(pred_file)

    # True labels
    if "true_label" in df.columns:
        y_true = df["true_label"].values
    else:
        y_true = df["label"].values

    # Probability of positive class
    if "probability_lung_opacity" in df.columns:
        p = df["probability_lung_opacity"].values
    elif "probability" in df.columns:
        p = df["probability"].values
    else:
        raise KeyError(
            f"{condition}: probability column not found."
        )

    # Brier score
    brier = brier_score_loss(
        y_true,
        p
    )

    # ECE
    ece, bin_df = expected_calibration_error(
        y_true,
        p,
        n_bins=10
    )

    # -----------------------------------------------------------------
    # Calibration curve
    # -----------------------------------------------------------------

    frac_pos, mean_pred = calibration_curve(
        y_true,
        p,
        n_bins=10,
        strategy="uniform"
    )

    plt.figure(figsize=(7, 6))

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Perfect calibration"
    )

    plt.plot(
        mean_pred,
        frac_pos,
        marker="o",
        label=condition
    )

    plt.xlabel("Mean Predicted Probability")
    plt.ylabel("Observed Frequency")
    plt.title(
        f"Calibration Curve — {condition}"
    )

    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()

    figure_path = os.path.join(
        OUTPUT_DIR,
        f"calibration_{condition}.png"
    )

    plt.savefig(
        figure_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    # -----------------------------------------------------------------
    # Print
    # -----------------------------------------------------------------

    print(
        f"Brier Score : {brier:.6f}"
    )

    print(
        f"ECE         : {ece:.6f}"
    )

    # -----------------------------------------------------------------
    # Save bins
    # -----------------------------------------------------------------

    bin_file = os.path.join(
        OUTPUT_DIR,
        f"calibration_bins_{condition}.csv"
    )

    bin_df.to_csv(
        bin_file,
        index=False
    )

    summary.append({
        "condition": condition,
        "brier_score": brier,
        "ECE": ece
    })


# ---------------------------------------------------------------------
# FINAL SUMMARY
# ---------------------------------------------------------------------

summary_df = pd.DataFrame(summary)

summary_file = os.path.join(
    OUTPUT_DIR,
    "calibration_summary.csv"
)

summary_df.to_csv(
    summary_file,
    index=False
)

print("\n" + "=" * 75)
print("CALIBRATION SUMMARY")
print("=" * 75)

print(
    summary_df.to_string(index=False)
)

print("\nSaved:")
print(summary_file)

print("\nFigures:")
print(OUTPUT_DIR)

print("\n" + "=" * 75)
print("CALIBRATION ANALYSIS COMPLETE")
print("=" * 75)
```

```python
# =====================================================================
# BOOTSTRAP 95% CONFIDENCE INTERVALS
# Frozen test evaluation: Confidence vs Reliability
# =====================================================================

import os
import numpy as np
import pandas as pd

from sklearn.metrics import roc_auc_score

print("=" * 75)
print("BOOTSTRAP 95% CONFIDENCE INTERVALS")
print("Confidence vs Frozen Reliability")
print("=" * 75)

# ---------------------------------------------------------------------
# PATHS
# ---------------------------------------------------------------------

BASE_DIR = r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset\research_dataset"

QUALITY_DIR = os.path.join(
    BASE_DIR,
    "quality_analysis"
)

PRED_DIR = os.path.join(
    BASE_DIR,
    "degradation_test_predictions"
)

OUTPUT_FILE = os.path.join(
    QUALITY_DIR,
    "bootstrap_confidence_intervals.csv"
)

# ---------------------------------------------------------------------
# LOAD FROZEN RELIABILITY MODEL
# ---------------------------------------------------------------------

import joblib

MODEL_PATH = os.path.join(
    QUALITY_DIR,
    "reliability_model.pkl"
)

reliability_model = joblib.load(MODEL_PATH)

print("\nFrozen reliability model loaded.")

# ---------------------------------------------------------------------
# CONDITIONS
# ---------------------------------------------------------------------

conditions = {
    "clean": (
        os.path.join(
            BASE_DIR,
            "clean_test_predictions.csv"
        ),
        os.path.join(
            QUALITY_DIR,
            "clean_quality.csv"
        )
    ),
    "blur": (
        os.path.join(
            PRED_DIR,
            "blur_predictions.csv"
        ),
        os.path.join(
            QUALITY_DIR,
            "blur_quality.csv"
        )
    ),
    "noise": (
        os.path.join(
            PRED_DIR,
            "noise_predictions.csv"
        ),
        os.path.join(
            QUALITY_DIR,
            "noise_quality.csv"
        )
    ),
    "low_resolution": (
        os.path.join(
            PRED_DIR,
            "low_resolution_predictions.csv"
        ),
        os.path.join(
            QUALITY_DIR,
            "low_resolution_quality.csv"
        )
    ),
    "jpeg": (
        os.path.join(
            PRED_DIR,
            "jpeg_predictions.csv"
        ),
        os.path.join(
            QUALITY_DIR,
            "jpeg_quality.csv"
        )
    ),
    "brightness_contrast": (
        os.path.join(
            PRED_DIR,
            "brightness_contrast_predictions.csv"
        ),
        os.path.join(
            QUALITY_DIR,
            "brightness_contrast_quality.csv"
        )
    )
}

# ---------------------------------------------------------------------
# RISK-COVERAGE
# ---------------------------------------------------------------------

def calculate_risk_coverage(score, errors):

    order = np.argsort(-score)

    sorted_errors = errors[order]

    cumulative_errors = np.cumsum(
        sorted_errors
    )

    n = len(errors)

    coverage = (
        np.arange(1, n + 1) / n
    )

    risk = (
        cumulative_errors /
        np.arange(1, n + 1)
    )

    return coverage, risk


def calculate_aurc(score, errors):

    coverage, risk = calculate_risk_coverage(
        score,
        errors
    )

    x = np.concatenate(
        [[0], coverage]
    )

    y = np.concatenate(
        [[0], risk]
    )

    return np.trapezoid(y, x)


# ---------------------------------------------------------------------
# BOOTSTRAP FUNCTION
# ---------------------------------------------------------------------

def bootstrap_metrics(
    y_true,
    errors,
    confidence,
    reliability,
    n_bootstrap=2000,
    random_state=42
):

    rng = np.random.default_rng(
        random_state
    )

    n = len(y_true)

    confidence_auc = []
    reliability_auc = []

    confidence_aurc = []
    reliability_aurc = []

    delta_auc = []
    delta_aurc = []

    attempts = 0

    while len(confidence_auc) < n_bootstrap:

        attempts += 1

        indices = rng.integers(
            0,
            n,
            size=n
        )

        y_boot = y_true[indices]

        # AUROC requires both classes
        if len(np.unique(y_boot)) < 2:
            continue

        error_boot = errors[indices]

        conf_boot = confidence[indices]
        rel_boot = reliability[indices]

        # -------------------------------------------------------------
        # AUROC
        # -------------------------------------------------------------

        auc_conf = roc_auc_score(
            error_boot,
            1 - conf_boot
        )

        auc_rel = roc_auc_score(
            error_boot,
            1 - rel_boot
        )

        # -------------------------------------------------------------
        # AURC
        # -------------------------------------------------------------

        aurc_conf = calculate_aurc(
            conf_boot,
            error_boot
        )

        aurc_rel = calculate_aurc(
            rel_boot,
            error_boot
        )

        # -------------------------------------------------------------
        # STORE
        # -------------------------------------------------------------

        confidence_auc.append(
            auc_conf
        )

        reliability_auc.append(
            auc_rel
        )

        confidence_aurc.append(
            aurc_conf
        )

        reliability_aurc.append(
            aurc_rel
        )

        delta_auc.append(
            auc_rel - auc_conf
        )

        delta_aurc.append(
            aurc_rel - aurc_conf
        )

    return {
        "confidence_auc": np.array(
            confidence_auc
        ),
        "reliability_auc": np.array(
            reliability_auc
        ),
        "confidence_aurc": np.array(
            confidence_aurc
        ),
        "reliability_aurc": np.array(
            reliability_aurc
        ),
        "delta_auc": np.array(
            delta_auc
        ),
        "delta_aurc": np.array(
            delta_aurc
        )
    }


# ---------------------------------------------------------------------
# PROCESS EACH CONDITION
# ---------------------------------------------------------------------

results = []

for condition, (pred_file, quality_file) in conditions.items():

    print("\n" + "-" * 75)
    print(f"CONDITION: {condition.upper()}")
    print("-" * 75)

    pred_df = pd.read_csv(
        pred_file
    )

    quality_df = pd.read_csv(
        quality_file
    )

    # -------------------------------------------------------------
    # ALIGN
    # -------------------------------------------------------------

    if (
        "path" in pred_df.columns
        and "path" in quality_df.columns
    ):

        merged = pred_df.merge(
            quality_df[
                [
                    "path",
                    "sharpness",
                    "contrast",
                    "entropy",
                    "high_frequency_variation"
                ]
            ],
            on="path",
            how="inner"
        )

        alignment = "path"

    else:

        if len(pred_df) != len(quality_df):
            raise ValueError(
                f"{condition}: length mismatch."
            )

        merged = pred_df.copy()

        for col in [
            "sharpness",
            "contrast",
            "entropy",
            "high_frequency_variation"
        ]:

            merged[col] = (
                quality_df[col].values
            )

        alignment = "row_order"

    print(
        "Samples:",
        len(merged)
    )

    print(
        "Alignment:",
        alignment
    )

    # -------------------------------------------------------------
    # LABELS
    # -------------------------------------------------------------

    if "true_label" in merged.columns:
        y_true = merged[
            "true_label"
        ].values
    else:
        y_true = merged[
            "label"
        ].values

    # -------------------------------------------------------------
    # PREDICTION ERROR
    # -------------------------------------------------------------

    if "predicted_label" in merged.columns:
        predicted = merged[
            "predicted_label"
        ].values
    else:
        predicted = merged[
            "prediction"
        ].values

    errors = (
        predicted != y_true
    ).astype(int)

    # -------------------------------------------------------------
    # CONFIDENCE
    # -------------------------------------------------------------

    if (
        "probability_lung_opacity"
        in merged.columns
    ):

        probability = merged[
            "probability_lung_opacity"
        ].values

    elif "probability" in merged.columns:

        probability = merged[
            "probability"
        ].values

    else:

        raise KeyError(
            f"{condition}: probability column missing."
        )

    confidence = np.maximum(
        probability,
        1 - probability
    )

    # -------------------------------------------------------------
    # FROZEN RELIABILITY
    # -------------------------------------------------------------

    reliability_features = pd.DataFrame({

        "confidence": confidence,

        "sharpness":
            merged["sharpness"].values,

        "contrast":
            merged["contrast"].values,

        "entropy":
            merged["entropy"].values,

        "high_frequency_variation":
            merged[
                "high_frequency_variation"
            ].values
    })

    predicted_error_probability = (
        reliability_model.predict_proba(
            reliability_features
        )[:, 1]
    )

    reliability = (
        1 -
        predicted_error_probability
    )

    # -------------------------------------------------------------
    # ORIGINAL METRICS
    # -------------------------------------------------------------

    confidence_auc_original = roc_auc_score(
        errors,
        1 - confidence
    )

    reliability_auc_original = roc_auc_score(
        errors,
        1 - reliability
    )

    confidence_aurc_original = calculate_aurc(
        confidence,
        errors
    )

    reliability_aurc_original = calculate_aurc(
        reliability,
        errors
    )

    print(
        f"\nConfidence AUROC : "
        f"{confidence_auc_original:.4f}"
    )

    print(
        f"Reliability AUROC: "
        f"{reliability_auc_original:.4f}"
    )

    print(
        f"Confidence AURC  : "
        f"{confidence_aurc_original:.6f}"
    )

    print(
        f"Reliability AURC : "
        f"{reliability_aurc_original:.6f}"
    )

    # -------------------------------------------------------------
    # BOOTSTRAP
    # -------------------------------------------------------------

    print(
        "\nRunning 2000 bootstrap samples..."
    )

    boot = bootstrap_metrics(
        y_true=y_true,
        errors=errors,
        confidence=confidence,
        reliability=reliability,
        n_bootstrap=2000,
        random_state=42
    )

    # -------------------------------------------------------------
    # CI FUNCTION
    # -------------------------------------------------------------

    def ci(values):

        return (
            np.percentile(values, 2.5),
            np.percentile(values, 97.5)
        )

    conf_auc_low, conf_auc_high = ci(
        boot["confidence_auc"]
    )

    rel_auc_low, rel_auc_high = ci(
        boot["reliability_auc"]
    )

    conf_aurc_low, conf_aurc_high = ci(
        boot["confidence_aurc"]
    )

    rel_aurc_low, rel_aurc_high = ci(
        boot["reliability_aurc"]
    )

    delta_auc_low, delta_auc_high = ci(
        boot["delta_auc"]
    )

    delta_aurc_low, delta_aurc_high = ci(
        boot["delta_aurc"]
    )

    # -------------------------------------------------------------
    # PRINT
    # -------------------------------------------------------------

    print("\n95% Confidence Intervals:")

    print(
        f"Confidence AUROC : "
        f"{confidence_auc_original:.4f} "
        f"[{conf_auc_low:.4f}, "
        f"{conf_auc_high:.4f}]"
    )

    print(
        f"Reliability AUROC: "
        f"{reliability_auc_original:.4f} "
        f"[{rel_auc_low:.4f}, "
        f"{rel_auc_high:.4f}]"
    )

    print(
        f"Confidence AURC  : "
        f"{confidence_aurc_original:.6f} "
        f"[{conf_aurc_low:.6f}, "
        f"{conf_aurc_high:.6f}]"
    )

    print(
        f"Reliability AURC : "
        f"{reliability_aurc_original:.6f} "
        f"[{rel_aurc_low:.6f}, "
        f"{rel_aurc_high:.6f}]"
    )

    print(
        f"\nΔ AUROC "
        f"(Reliability - Confidence): "
        f"{delta_auc_low:+.4f} "
        f"to "
        f"{delta_auc_high:+.4f}"
    )

    print(
        f"Δ AURC "
        f"(Reliability - Confidence): "
        f"{delta_aurc_low:+.6f} "
        f"to "
        f"{delta_aurc_high:+.6f}"
    )

    # -------------------------------------------------------------
    # SAVE
    # -------------------------------------------------------------

    results.append({

        "condition": condition,

        "confidence_AUROC":
            confidence_auc_original,

        "confidence_AUROC_CI_low":
            conf_auc_low,

        "confidence_AUROC_CI_high":
            conf_auc_high,

        "reliability_AUROC":
            reliability_auc_original,

        "reliability_AUROC_CI_low":
            rel_auc_low,

        "reliability_AUROC_CI_high":
            rel_auc_high,

        "confidence_AURC":
            confidence_aurc_original,

        "confidence_AURC_CI_low":
            conf_aurc_low,

        "confidence_AURC_CI_high":
            conf_aurc_high,

        "reliability_AURC":
            reliability_aurc_original,

        "reliability_AURC_CI_low":
            rel_aurc_low,

        "reliability_AURC_CI_high":
            rel_aurc_high,

        "delta_AUROC":
            reliability_auc_original -
            confidence_auc_original,

        "delta_AUROC_CI_low":
            delta_auc_low,

        "delta_AUROC_CI_high":
            delta_auc_high,

        "delta_AURC":
            reliability_aurc_original -
            confidence_aurc_original,

        "delta_AURC_CI_low":
            delta_aurc_low,

        "delta_AURC_CI_high":
            delta_aurc_high
    })


# ---------------------------------------------------------------------
# FINAL TABLE
# ---------------------------------------------------------------------

results_df = pd.DataFrame(
    results
)

results_df.to_csv(
    OUTPUT_FILE,
    index=False
)

print("\n" + "=" * 75)
print("FINAL BOOTSTRAP RESULTS")
print("=" * 75)

print(
    results_df.to_string(
        index=False
    )
)

print("\nSaved:")
print(OUTPUT_FILE)

print("\n" + "=" * 75)
print("BOOTSTRAP ANALYSIS COMPLETE")
print("=" * 75)
```

```python
# =============================================================================
# FIND ECE / CALIBRATION CODE + RESULTS
# =============================================================================

import os
import re
import pandas as pd

ROOT = r"C:\Users\mohsi\Downloads\covid_radiography_dataset\COVID-19_Radiography_Dataset\research_dataset"

print("=" * 90)
print("SEARCHING PROJECT FOR ECE / CALIBRATION CODE")
print("=" * 90)

# -------------------------------------------------------------------------
# 1. Search all text/code files for ECE / calibration references
# -------------------------------------------------------------------------

patterns = [
    r"\bECE\b",
    r"expected_calibration_error",
    r"calibration",
    r"n_bins",
    r"num_bins",
    r"bins\s*=",
    r"calibration_curve"
]

found = []

for root, dirs, files in os.walk(ROOT):
    # Skip huge/generated directories
    dirs[:] = [
        d for d in dirs
        if d not in {
            ".git",
            "__pycache__",
            "degradation_test_sets"
        }
    ]

    for filename in files:
        path = os.path.join(root, filename)

        if not filename.lower().endswith((
            ".py", ".ipynb", ".txt", ".md", ".csv"
        )):
            continue

        try:
            if filename.lower().endswith(".ipynb"):
                import json

                with open(path, "r", encoding="utf-8") as f:
                    nb = json.load(f)

                for i, cell in enumerate(nb.get("cells", [])):
                    source = "".join(cell.get("source", []))

                    if any(re.search(pattern, source, re.I) for pattern in patterns):
                        found.append((path, f"cell {i}", source))

            else:
                with open(path, "r", encoding="utf-8", errors="ignore") as f:
                    source = f.read()

                if any(re.search(pattern, source, re.I) for pattern in patterns):
                    found.append((path, "file", source))

        except Exception:
            pass


# -------------------------------------------------------------------------
# 2. Print relevant code snippets
# -------------------------------------------------------------------------

print("\n" + "=" * 90)
print(f"FILES / CELLS CONTAINING CALIBRATION REFERENCES: {len(found)}")
print("=" * 90)

for idx, (path, location, source) in enumerate(found, 1):

    print(f"\n{'-' * 90}")
    print(f"[{idx}] {path}")
    print(f"Location: {location}")
    print(f"{'-' * 90}")

    lines = source.splitlines()

    relevant = []

    for i, line in enumerate(lines):
        if any(re.search(pattern, line, re.I) for pattern in patterns):

            # Show surrounding lines
            start = max(0, i - 5)
            end = min(len(lines), i + 8)

            for j in range(start, end):
                relevant.append((j + 1, lines[j]))

            relevant.append(("---", "---"))

    # Remove duplicates while preserving order
    seen = set()

    for line_no, line in relevant:
        key = (line_no, line)

        if key not in seen:
            print(f"{str(line_no):>5}: {line}")
            seen.add(key)


# -------------------------------------------------------------------------
# 3. Load calibration summary
# -------------------------------------------------------------------------

print("\n" + "=" * 90)
print("CALIBRATION SUMMARY")
print("=" * 90)

calibration_files = []

for root, dirs, files in os.walk(ROOT):
    for filename in files:
        if filename.lower() == "calibration_summary.csv":
            calibration_files.append(os.path.join(root, filename))

if calibration_files:

    for path in calibration_files:

        print(f"\nFile: {path}")

        try:
            cal = pd.read_csv(path)

            print("\nColumns:")
            print(cal.columns.tolist())

            print("\nResults:")
            print(cal.to_string(index=False))

        except Exception as e:
            print("Could not read:", e)

else:
    print("No calibration_summary.csv found.")


# -------------------------------------------------------------------------
# 4. Search CSVs for ECE values
# -------------------------------------------------------------------------

print("\n" + "=" * 90)
print("CSV FILES CONTAINING ECE")
print("=" * 90)

for root, dirs, files in os.walk(ROOT):

    for filename in files:

        if not filename.lower().endswith(".csv"):
            continue

        path = os.path.join(root, filename)

        try:
            df = pd.read_csv(path, nrows=5)

            ece_cols = [
                c for c in df.columns
                if "ece" in str(c).lower()
            ]

            if ece_cols:
                print(f"\n{path}")
                print("ECE columns:", ece_cols)

        except Exception:
            pass


# -------------------------------------------------------------------------
# 5. Explicitly search for bin-count values
# -------------------------------------------------------------------------

print("\n" + "=" * 90)
print("POSSIBLE ECE BIN COUNTS")
print("=" * 90)

bin_patterns = [
    r"n_bins\s*=\s*(\d+)",
    r"num_bins\s*=\s*(\d+)",
    r"bins\s*=\s*(\d+)",
    r"range\s*\(\s*(\d+)\s*\)"
]

for path, location, source in found:

    for pattern in bin_patterns:

        matches = re.findall(pattern, source, re.I)

        if matches:

            print(
                f"{os.path.basename(path)} | {location} | "
                f"{pattern} -> {matches}"
            )


print("\n" + "=" * 90)
print("DONE")
print("=" * 90)
```

```python
# =============================================================================
# SEARCH ALL JUPYTER NOTEBOOKS FOR CALIBRATION / ECE CODE
# =============================================================================

import os
import json
import re

SEARCH_ROOTS = [
    r"C:\Users\mohsi\Downloads",
    r"C:\Users\mohsi"
]

patterns = [
    "ECE",
    "expected_calibration",
    "calibration_curve",
    "brier_score",
    "n_bins",
    "num_bins",
    "bin_edges",
    "reliability"
]

seen_files = set()
results = []

for search_root in SEARCH_ROOTS:

    if not os.path.exists(search_root):
        continue

    for root, dirs, files in os.walk(search_root):

        # Skip directories that can contain huge numbers of irrelevant files
        dirs[:] = [
            d for d in dirs
            if d not in {
                ".git",
                "node_modules",
                "__pycache__",
                "site-packages",
                "AppData",
                "Program Files",
                "Program Files (x86)"
            }
        ]

        for filename in files:

            if not filename.lower().endswith(".ipynb"):
                continue

            path = os.path.abspath(os.path.join(root, filename))

            if path in seen_files:
                continue

            seen_files.add(path)

            try:
                with open(path, "r", encoding="utf-8") as f:
                    notebook = json.load(f)

                for cell_idx, cell in enumerate(notebook.get("cells", [])):

                    source = "".join(cell.get("source", []))

                    if any(
                        keyword.lower() in source.lower()
                        for keyword in patterns
                    ):
                        results.append(
                            (path, cell_idx, source)
                        )

            except Exception:
                pass


print("=" * 100)
print("CALIBRATION NOTEBOOK SEARCH")
print("=" * 100)

print(f"\nNotebooks scanned: {len(seen_files)}")
print(f"Matching cells found: {len(results)}")


for i, (path, cell_idx, source) in enumerate(results, 1):

    print("\n" + "-" * 100)
    print(f"[MATCH {i}]")
    print("Notebook :", path)
    print("Cell     :", cell_idx)
    print("-" * 100)

    lines = source.splitlines()

    # Print complete matching cell
    for line_no, line in enumerate(lines, 1):
        print(f"{line_no:4}: {line}")


print("\n" + "=" * 100)
print("DONE")
print("=" * 100)
```
